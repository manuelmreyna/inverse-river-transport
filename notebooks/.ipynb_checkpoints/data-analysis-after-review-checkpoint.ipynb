```python
ts_m_0 = []
btcs_m_0 = []
for i_btc in range(n_btcs):
    t,btc = ts_m[i_btc],btcs_m[i_btc]
    sorted_indices = np.argsort(t)
    t = t[sorted_indices]
    btc = btc[sorted_indices]
    t_unique,indices_unique = np.unique(t, return_index=True)
    if len(t_unique)!=len(t):
        btc_unique = np.zeros(len(indices_unique))
        for i in range(len(indices_unique)):
            if i<len(indices_unique)-1:
                btc_unique[i] = np.mean(btc[indices_unique[i]:indices_unique[i+1]])
            else:
                btc_unique[i] = np.mean(btc[indices_unique[i]:])
        t = t_unique
        btc = btc_unique
    deltat = np.median(t[1:]-t[:-1])
    t_ext = np.concatenate((np.arange(t[0],0,-deltat)[1:][::-1],t,np.arange(t[-1],max(t[-1],T_*t[np.argmax(btc)]),deltat)[1:]))
    btc_ext = np.interp(t_ext,t,btc-background_cs_m[i_btc],left = 0.0,right = 0.0)
    btc_ext = btc_ext
    ts_m_0.append(t_ext)
    btcs_m_0.append(btc_ext)
```

```python
def plot_test_norm_log(prop,name_prop,ax):    
    distances_normalized = np.array([])
    prop_normalized = np.array([])
    for ind_exp in index_experiments:
        ax.scatter((xs_m)[ind_exp]/np.exp(np.mean(np.log((xs_m)[ind_exp]))),prop[ind_exp]/np.exp(np.mean(np.log(prop[ind_exp]))),linewidth =1, c = 'tab:blue',alpha = 0.5)
        distances_normalized = np.concatenate((distances_normalized,(xs_m)[ind_exp]/np.exp(np.mean(np.log((xs_m)[ind_exp])))))
        prop_normalized = np.concatenate((prop_normalized,(prop)[ind_exp]/np.exp(np.mean(np.log((prop)[ind_exp])))))
    
    
    m,b = np.polyfit(np.log(distances_normalized[~np.isnan(prop_normalized)]), np.log(prop_normalized[~np.isnan(prop_normalized)]), 1)
    print(m,np.exp(b))
    logR2 = 1-np.sum((np.log(prop_normalized[prop_normalized>0])-(b+m*np.log(distances_normalized[prop_normalized>0])))**2)/np.sum((np.log(prop_normalized[prop_normalized>0])-np.mean(np.log(prop_normalized[prop_normalized>0])))**2)
    ax.plot(np.array([np.min(distances_normalized),np.max(distances_normalized)]),np.exp(b)*np.array([np.min(distances_normalized),np.max(distances_normalized)])**m,c = 'k', linestyle = 'dashed', label = r'$ \left(\frac{L}{\bar{L}} \right)^{'+format(m,'.3')+'}$, $R_{log}^2='+format(logR2,'.3')+'$')
    
    ax.grid(which='major')
    ax.grid(which='minor',linewidth = 0.3)
    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.set_xlabel(r'$L/\bar{L}$')
    ax.set_ylabel(r'$'+name_prop+'/\overline{'+name_prop+'}$')
    ax.legend()
```

```python
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b', '#e377c2', '#7f7f7f', '#bcbd22', '#17becf']
```

```python
memory_func = 'first order'
bound_cond = 'infinite'
only_params = False
```

```python
vs_peak = np.array([x/ts_meas[np.argmax(measured_bt)] for ts_meas,measured_bt,x in zip(ts_m_0,btcs_m_0,xs_m)])
ts_peak = np.array([ts_meas[np.argmax(measured_bt)] for ts_meas,measured_bt,x in zip(ts_m_0,btcs_m_0,xs_m)])
```

```python
%%time
def estimate_laplace(i_btc):
    t,btc,x, v_ref = ts_m_0[i_btc],btcs_m_0[i_btc],xs_m[i_btc], vs_peak[i_btc]
    t_pos = t[btc>0]
    btc_pos = btc[btc>0]
    params0 = np.array([vs_peak[i_btc],100.0,3.0,30.0])
    params = estimators.laplace.ADEMF_1D(t_pos,btc_pos,x,'first order','infinite',params0)
    if t[0]==0:
        btc_rec = np.concatenate(([0.0],forward.laplace.ADEMF_1D(t[1:],[x],params[0],min(params[1],20000),params[2]/params[3],[params[3],params[3]],'first order','infinite')[:,-1]))
    else:
        btc_rec = forward.laplace.ADEMF_1D(t,[x],params[0],min(params[1],20000),params[2]/params[3],[params[3],params[3]],'first order','infinite')[:,-1]
    RMSD = utils.errors.RMSD_func(t,btc,btc_rec)
    KL_div = utils.errors.KL_div_func(t,btc,btc_rec)
    return params,RMSD,KL_div,btc_rec
output_estimate_laplace = [estimate_laplace(i_btc) for i_btc in range(n_btcs)]
params_laplace = np.array([out[0] for out in output_estimate_laplace])
RMSDs_laplace = np.array([out[1] for out in output_estimate_laplace])
KL_divs_laplace = np.array([out[2] for out in output_estimate_laplace])
btcs_rec_laplace = [out[3] for out in output_estimate_laplace]
```

```python
%%time
if memory_func == 'first order':
    def estimate_moments(i_btc):
        t,btc,x = ts_m_0[i_btc],btcs_m_0[i_btc],xs_m[i_btc]
        v_prior = x/t[np.argmax(btc)]
        params0 =  np.array([v_prior,300.0,3.0,30.0])
        params = estimators.moments.ADEFO_1D(t,btc,x,'infinite',params0)
        if t[0]==0:
            btc_rec = np.concatenate(([0.0],forward.laplace.ADEMF_1D(t[1:],[x],params[0],min(params[1],20000),params[2]/params[3],[params[3],params[3]],'first order','infinite')[:,-1]))
        else:
            btc_rec = forward.laplace.ADEMF_1D(t,[x],params[0],min(params[1],20000),params[2]/params[3],[params[3],params[3]],'first order','infinite')[:,-1]
        RMSD = utils.errors.RMSD_func(t,btc,btc_rec)
        KL_div = utils.errors.KL_div_func(t,btc,btc_rec)
        return params,RMSD,KL_div,btc_rec
    output_estimate_moments = [estimate_moments(i_btc) for i_btc in range(n_btcs)]
    params_moments = np.array([out[0] for out in output_estimate_moments])
    RMSDs_moments = np.array([out[1] for out in output_estimate_moments])
    KL_divs_moments = np.array([out[2] for out in output_estimate_moments])
    btcs_rec_moments = [out[3] for out in output_estimate_moments]
```

```python
from matplotlib.transforms import offset_copy
# np.random.seed(1)
Nrows = 3
Ncols = 4
fig,axes = plt.subplots(nrows = Nrows, ncols = Ncols, figsize = (12,6))
np.random.seed(42)
rand_i_btcs = np.sort(np.random.choice(list(range(n_btcs)),size = Nrows*Ncols, replace = False))
j=0
for i_btc in rand_i_btcs:
    t,btc,x,est_btc_lap = ts_m_0[i_btc],btcs_m_0[i_btc],xs_m[i_btc],btcs_rec_laplace[i_btc]
    axes[j//Ncols][j%Ncols].scatter(t,btc, c='k', marker = 'x',label = '$\epsilon$=0')
    axes[j//Ncols][j%Ncols].set_ylabel('c', rotation=0)
    axes[j//Ncols][j%Ncols].set_xlabel('t', loc='right', labelpad=-10)
    if memory_func == 'first order':
        est_btc_mom = btcs_rec_moments[i_btc]
        axes[j//Ncols][j%Ncols].plot(t, est_btc_mom*np.trapezoid(btc,t)/np.trapezoid(est_btc_mom,t),label = r'$\epsilon$='+format(RMSDs_moments[i_btc],'.0%'), c = colors[0])
    
    axes[j//Ncols][j%Ncols].plot(t, est_btc_lap*np.trapezoid(btc,t)/np.trapezoid(est_btc_lap,t),label = r'$\epsilon$='+format(RMSDs_laplace[i_btc],'.0%'), c = colors[1])
    
    axes[j//Ncols][j%Ncols].set_xlim(t[btc>np.max(btc)*0.001][0],t[btc>np.max(btc)*0.01][-1])
    axes[j//Ncols][j%Ncols].set_ylim(0.0,np.max(btc))
    # axes[j//Ncols][j%Ncols].set_title(names[i_btc], loc='right',pad = -50)
    axes[j//Ncols][j%Ncols].legend()

    axes[j//Ncols][j%Ncols].grid()
    axes[j//Ncols][j%Ncols].yaxis.set_label_coords(-0.04,1.01)
    axes[j//Ncols][j%Ncols].xaxis.set_label_coords(1.02,-0.03)

    j+=1
handles, _ = axes[-1][-1].get_legend_handles_labels()
if memory_func == 'first order':
    labels = ['Measured','Moment matching', 'Laplace domain fitting']#, 'KL-DNN-MAP']
else:
    labels = ['Measured', 'Laplace domain fitting', 'KL-DNN-MAP']
fig.legend(handles, labels, loc=(0.15,0.0015), ncols = 3)
```

```python
pars_index = [0,1]
# 1️⃣ Eigen-decomposition
eigvals, eigvecs = np.linalg.eigh(4*cov_log_params[pars_index][:,pars_index])

# 2️⃣ Parameterize unit circle
theta = np.linspace(0, 2*np.pi, 200)
circle = np.vstack((np.cos(theta), np.sin(theta)))

# 3️⃣ Scale to ellipse in log-space
ellipse_log = (
    eigvecs @ np.diag(np.sqrt(eigvals)) @ circle
    + mean_log_params[pars_index][:,None]
)

# 4️⃣ Transform back to original space
ellipse = np.exp(ellipse_log)

# 5️⃣ Plot
plt.figure()
plt.loglog(ellipse[0], ellipse[1], 'tab:orange', lw=2)
plt.loglog(ellipse[0], ellipse[1], 'tab:orange', lw=2)
plt.scatter(*dimless_best_params_prior[indices_train][~is_outlier][:,pars_index].T, c = np.log(np.minimum(RMSDs_laplace,RMSDs_moments)[indices_train][~is_outlier]), s = 2.0)
# plt.scatter(*dimless_best_params_prior[indices_train][is_outlier][:,pars_index].T, s = 2.0, c ='r')
plt.colorbar()
plt.xscale('log')
plt.yscale('log')
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
def RMSE_t_scale(i_btc,t_scale,ts_synth,synth_btc):
    t_peak_synth = ts_synth[np.argmax(synth_btc)]
    btc_interp = np.interp(ts_m_0[i_btc],ts_synth*t_scale/t_peak_synth,synth_btc)
    a_fac = np.trapezoid(btc_interp*btcs_m_0[i_btc],x = ts_m_0[i_btc])/np.trapezoid(btc_interp**2,x = ts_m_0[i_btc])
    btc_interp_scaled = btc_interp*a_fac
    RMSD = np.sqrt(np.trapezoid((btcs_m_0[i_btc]-btc_interp_scaled)**2,x = ts_m_0[i_btc])/np.trapezoid((btcs_m_0[i_btc])**2,x = ts_m_0[i_btc]))
    return RMSD,btc_interp_scaled,a_fac
```

```python
%%time
from scipy.optimize import differential_evolution
n_iter_DE = 50
est_params_DE = np.zeros((n_btcs,3))
nfev_DE = np.zeros(n_btcs)
histories_DE = np.zeros((n_btcs,1000))
histories_params_DE = np.zeros((n_btcs,1000,3))
Nt = 150
ts_synth = np.linspace(0,T_,T_*Nt+1)[1:]
for i_btc in range(n_btcs):
    print(i_btc)
    history = []
    history_params = []
    cov_log_params_sqrt_3 = sp.linalg.sqrtm(cov_log_params[1:,1:])
    range_t_scale_rel_val = 0.8
    index_max_first,index_max_last = np.where(btcs_m_0[i_btc]>range_t_scale_rel_val*np.max(btcs_m_0[i_btc]))[0][[0,-1]]
    def interp_rel_val(index):
        return (ts_m_0[i_btc][index-1]
                +(ts_m_0[i_btc][index]-ts_m_0[i_btc][index-1])
                /(btcs_m_0[i_btc][index]/np.max(btcs_m_0[i_btc])-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc]))
                *(range_t_scale_rel_val-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc])))
    range_t_scale = [interp_rel_val(index_max_first),interp_rel_val(index_max_last+1)]
    
    def loss(phi):
        theta = np.exp(cov_log_params_sqrt_3@sp.stats.norm.ppf(phi)+mean_log_params[1:])
        btc_synth = forward.laplace.ADEMF_1D(ts_synth,[1.0],1.0,theta[0],theta[1]/theta[2],[theta[2],theta[2]],'first order','infinite')[:,-1]
        output = sp.optimize.minimize_scalar(lambda x: RMSE_t_scale(i_btc,x,ts_synth,btc_synth)[0], method='Brent', bracket=range_t_scale)
        loss = output.fun
        history.append(loss)
        history_params.append(theta)
        return loss
    
    bounds = [(0.01,0.99) for i in range(3)]
    
    # 3. Run DE with explicit arguments
    de_result = differential_evolution(
        func=loss, 
        bounds=bounds, 
        strategy='best1bin', 
        maxiter=n_iter_DE, 
        popsize=6, 
        recombination=0.7, 
        tol=0,          # disable convergence criterion
        atol=0,
        seed=global_seed#,
        # callback = callback
    )

    

    histories_DE[i_btc][:min(len(history),1000)] = history[:min(len(history),1000)]
    histories_params_DE[i_btc][:min(len(history),1000)] = history_params[:min(len(history),1000)]
    est_params_DE[i_btc] = np.exp(cov_log_params_sqrt_3@sp.stats.norm.ppf(de_result.x)+mean_log_params[1:])
    nfev_DE[i_btc] = de_result.nfev
    print(de_result.success)
```

```python

from scipy.spatial import Delaunay, cKDTree

def linear_ND_interp(tri, tree, x):

    x = np.asarray(x)
    ndim = tri.ndim
    simplex = tri.find_simplex(x)

    if simplex >= 0:
        # Standard barycentric interpolation
        T = tri.transform[simplex, :ndim]
        r = tri.transform[simplex, ndim]

        bary = T @ (x - r)
        weights = np.concatenate([bary, [1 - bary.sum()]])
        vertices = tri.simplices[simplex]

    else:
        # Outside convex hull → nearest vertex
        dist, nearest = tree.query(x)

        vertices = np.zeros(ndim + 1, dtype=int)
        vertices[0] = nearest

        # fill remaining slots with any other indices
        remaining = np.setdiff1d(np.arange(len(tri.points)), [nearest])
        vertices[1:] = remaining[:ndim]

        weights = np.zeros(ndim + 1)
        weights[0] = 1.0
        # vertices, weights = np.zeros(ndim+1).astype(int), np.zeros(ndim+1)+np.nan

    return vertices, weights


def Zs_interpolate(pars,tri,tree,Zs):
    ND_interp_vertices,ND_interp_weights = linear_ND_interp(tri,tree,pars)
    return ND_interp_weights@Zs[ND_interp_vertices],ND_interp_weights@ts_peak_synth[ND_interp_vertices]
    
triangulation = Delaunay(np.log(rand_params_synth))
tree = cKDTree(np.log(rand_params_synth))
```

```python
%%time
RMSDs_NNI_full = np.zeros(n_btcs)
est_params_NNI_full = np.zeros((n_btcs,5))
btcs_recons_NNI_full = []
for i_btc in range(n_btcs):
    range_t_scale_rel_val = 0.8
    index_max_first,index_max_last = np.where(btcs_m_0[i_btc]>range_t_scale_rel_val*np.max(btcs_m_0[i_btc]))[0][[0,-1]]
    def interp_rel_val(index):
        return (ts_m_0[i_btc][index-1]
                +(ts_m_0[i_btc][index]-ts_m_0[i_btc][index-1])
                /(btcs_m_0[i_btc][index]/np.max(btcs_m_0[i_btc])-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc]))
                *(range_t_scale_rel_val-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc])))
    range_t_scale = [interp_rel_val(index_max_first),interp_rel_val(index_max_last+1)]
    outs_RMSE_t_scale = [sp.optimize.minimize_scalar(lambda x: RMSE_t_scale(i_btc,x,ts_synth,synth_btcs[i])[0], method='bounded', bounds=range_t_scale) for i in range(n_synth)]
    t_scales = [out.x for out in outs_RMSE_t_scale]
    RMSDs = [out.fun for out in outs_RMSE_t_scale]
    btcs_interp_scaled = [RMSE_t_scale(i_btc,t_scales[i],ts_synth,synth_btcs[i])[1] for i in range(n_synth)]
    a_facs_est = np.array([RMSE_t_scale(i_btc,t_scales[i],ts_synth,synth_btcs[i])[2] for i in range(n_synth)])
    RMSDs_NNI_full[i_btc] = np.min(RMSDs)
    NNI_index = np.argmin(RMSDs)
    vs_est = xs_m[i_btc]/(t_scales/ts_peak_synth)
    est_pars = np.concatenate(([a_facs_est[NNI_index],vs_est[NNI_index]],rand_params_synth[NNI_index]))
    est_params_NNI_full[i_btc] = est_pars
    if ts_m[i_btc][0]>0:
        btcs_recons_NNI_full.append(forward.laplace.ADEMF_1D(ts_m[i_btc],[xs_m[i_btc]],est_pars[1],est_pars[2],est_pars[3]/est_pars[4],[est_pars[4],est_pars[4]],'first order','infinite')[:,-1]*est_pars[0]/est_pars[1]*xs_m[i_btc])
    else:
        btcs_recons_NNI_full.append(np.concatenate(([0],forward.laplace.ADEMF_1D(ts_m[i_btc][1:],[xs_m[i_btc]],est_pars[1],est_pars[2],est_pars[3]/est_pars[4],[est_pars[4],est_pars[4]],'first order','infinite')[:,-1]*est_pars[0]/est_pars[1]*xs_m[i_btc])))
    print(i_btc,np.min(RMSDs))
```

```python
def RMSE_t_scale_Z(i_btc,t_scale,Zs_i,t_peak_synth_interp,btc_mean,phis):
    btc_mean_resampled = np.interp(ts_m_0[i_btc],ts_synth*t_scale/t_peak_synth_interp,btc_mean)
    phis_resampled = np.array([np.interp(ts_m_0[i_btc],ts_synth*t_scale/t_peak_synth_interp,phi) for phi in phis.T]).T
    btc_interp =  (btc_mean_resampled+Zs_i@(phis_resampled).T)
    a_fac = np.trapezoid(btc_interp*btcs_m_0[i_btc],x = ts_m_0[i_btc])/np.trapezoid(btc_interp**2,x = ts_m_0[i_btc])
    btc_interp_scaled = btc_interp*a_fac
    RMSD = np.sqrt(np.trapezoid((btcs_m_0[i_btc]-btc_interp_scaled)**2,x = ts_m_0[i_btc])/np.trapezoid((btcs_m_0[i_btc])**2,x = ts_m_0[i_btc]))
    return RMSD,btc_interp_scaled
```

```python
%%time
RMSDs_linear_full = np.zeros((n_btcs))
est_params_linear_full = np.zeros((n_btcs,5))
btcs_recons_linear_full = []

for i_btc in range(n_btcs):
    if len(btcs_m[i_btc])>N:
        weight_reg = 0.0001
    else:
        weight_reg = 0.001
    np.random.seed(global_seed)
    bounds = np.concatenate((np.array([-np.inf,np.log(5000)])[None,:],
                             np.array([-np.inf,np.inf])*np.ones(2)[:,None]))
    w = np.trapezoid(np.eye(len(ts_m_0[i_btc])),x= ts_m_0[i_btc])

    range_t_scale_rel_val = 0.8
    index_max_first,index_max_last = np.where(btcs_m_0[i_btc]>range_t_scale_rel_val*np.max(btcs_m_0[i_btc]))[0][[0,-1]]
    def interp_rel_val(index):
        return (ts_m_0[i_btc][index-1]
                +(ts_m_0[i_btc][index]-ts_m_0[i_btc][index-1])
                /(btcs_m_0[i_btc][index]/np.max(btcs_m_0[i_btc])-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc]))
                *(range_t_scale_rel_val-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc])))
    range_t_scale = [interp_rel_val(index_max_first),interp_rel_val(index_max_last+1)]


    def Zs_interpolate(pars,tri,tree,Zs):
        ND_interp_vertices,ND_interp_weights = linear_ND_interp(tri,tree,pars)
        return ND_interp_weights@Zs[ND_interp_vertices],ND_interp_weights@ts_peak_synth[ND_interp_vertices]
    
    triangulation = Delaunay(np.log(rand_params_synth))
    tree = cKDTree(np.log(rand_params_synth))
    
    RMSD_denom = np.sum(w*(btcs_m_0[i_btc])**2)
    def loss(par):
        Zs_i,t_peak_synth_interp = Zs_interpolate(par,triangulation,tree,Zs)
        return sp.optimize.minimize_scalar(lambda x: RMSE_t_scale_Z(i_btc,x,Zs_i,t_peak_synth_interp,
                btc_mean,phis)[0], method='bounded', bounds=range_t_scale).fun + weight_reg * np.sum(Zs_i**2/lambdas)
    results = [sp.optimize.minimize(loss,x0 = np.log(est_params_NNI_full[i_btc,2:]),method='Nelder-Mead',bounds = bounds)]
    RMSDs_interp = [result.fun for result in results]
    index = np.argmin(RMSDs_interp)
    est_linear = np.exp(results[index].x)
    btc_linear = forward.laplace.ADEMF_1D(ts_synth,[1.0],1.0,est_linear[0],est_linear[1]/est_linear[2],[est_linear[2],est_linear[2]],'first order','infinite')[:,-1]
    t_scale_linear = sp.optimize.minimize_scalar(lambda x: RMSE_t_scale(i_btc,x,ts_synth,btc_linear)[0], method='bounded', bounds=range_t_scale).x
    RMSD,btc_interp_scaled,a_fac = RMSE_t_scale(i_btc,t_scale_linear,ts_synth,btc_linear)
    v_est = xs_m[i_btc]/(t_scale_linear/ts_synth[np.argmax(btc_linear)])
    est_pars = [a_fac,v_est,*est_linear]
    est_params_linear_full[i_btc] = est_pars
    RMSDs_linear_full[i_btc] = min(RMSDs_NNI_full[i_btc],RMSD)
    if ts_m[i_btc][0]>0:
        btcs_recons_linear_full.append(forward.laplace.ADEMF_1D(ts_m_0[i_btc],[xs_m[i_btc]],est_pars[1],est_pars[2],est_pars[3]/est_pars[4],[est_pars[4],est_pars[4]],'first order','infinite')[:,-1]*est_pars[0]/est_pars[1]*xs_m[i_btc])
    else:
        btcs_recons_linear_full.append(np.concatenate(([0],forward.laplace.ADEMF_1D(ts_m_0[i_btc][1:],[xs_m[i_btc]],est_pars[1],est_pars[2],est_pars[3]/est_pars[4],[est_pars[4],est_pars[4]],'first order','infinite')[:,-1]*est_pars[0]/est_pars[1]*xs_m[i_btc])))
    print(i_btc,RMSDs_NNI_full[i_btc],RMSD,RMSDs_interp[index])
```

```python
%%time
N = 16
Zs_all = np.zeros((len(ns_synth_interm_linear),n_synth,N))
phis_all = np.zeros((len(ns_synth_interm_linear),len(ts_synth),N))
btc_mean_all = np.zeros((len(ns_synth_interm_linear),len(ts_synth)))
lambdas_all = np.zeros((len(ns_synth_interm_linear),N))
for j,n in enumerate(ns_synth_interm_linear):
    btc_mean = np.mean(synth_btcs[:n],axis = 0)
    btc_diff = synth_btcs[:n]-btc_mean
    C = btc_diff.T @ btc_diff / (btc_diff.shape[0] - 1)
    eigvals, eigvecs = np.linalg.eigh(C)
    idx = np.argsort(eigvals)[::-1]
    N_terms = min(N,n)
    lambdas = eigvals[idx][:N_terms]
    phis = eigvecs[:, idx][:,:N_terms]
    Zs = (btc_diff @ eigvecs[:, idx])[:,:N_terms]
    Zs_all[j,:n,:N_terms] = Zs
    phis_all[j,:,:N_terms] = phis
    btc_mean_all[j] = btc_mean
    lambdas_all[j,:N_terms] = lambdas
```

```python

from scipy.spatial import Delaunay, cKDTree

def linear_ND_interp(tri, tree, x):

    x = np.asarray(x)
    ndim = tri.ndim
    simplex = tri.find_simplex(x)

    if simplex >= 0:
        # Standard barycentric interpolation
        T = tri.transform[simplex, :ndim]
        r = tri.transform[simplex, ndim]

        bary = T @ (x - r)
        weights = np.concatenate([bary, [1 - bary.sum()]])
        vertices = tri.simplices[simplex]

    else:
        # Outside convex hull → nearest vertex
        dist, nearest = tree.query(x)

        vertices = np.zeros(ndim + 1, dtype=int)
        vertices[0] = nearest

        # fill remaining slots with any other indices
        remaining = np.setdiff1d(np.arange(len(tri.points)), [nearest])
        vertices[1:] = remaining[:ndim]

        weights = np.zeros(ndim + 1)
        weights[0] = 1.0
        # vertices, weights = np.zeros(ndim+1).astype(int), np.zeros(ndim+1)+np.nan

    return vertices, weights


def Zs_interpolate(pars,tri,tree,Zs):
    ND_interp_vertices,ND_interp_weights = linear_ND_interp(tri,tree,pars)
    return ND_interp_weights@Zs[ND_interp_vertices],ND_interp_weights@ts_peak_synth[ND_interp_vertices]
    
triangulation = Delaunay(np.log(rand_params_synth))
tree = cKDTree(np.log(rand_params_synth))
```

```python
%%time
RMSDs_NNI = np.zeros((n_btcs,n_synth))
est_params_NNI = np.zeros((n_btcs,n_synth,5))
for i_btc in range(n_btcs):
    range_t_scale_rel_val = 0.8
    index_max_first,index_max_last = np.where(btcs_m_0[i_btc]>range_t_scale_rel_val*np.max(btcs_m_0[i_btc]))[0][[0,-1]]
    def interp_rel_val(index):
        return (ts_m_0[i_btc][index-1]
                +(ts_m_0[i_btc][index]-ts_m_0[i_btc][index-1])
                /(btcs_m_0[i_btc][index]/np.max(btcs_m_0[i_btc])-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc]))
                *(range_t_scale_rel_val-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc])))
    range_t_scale = [interp_rel_val(index_max_first),interp_rel_val(index_max_last+1)]
    outs_RMSE_t_scale = [sp.optimize.minimize_scalar(lambda x: RMSE_t_scale(i_btc,x,ts_synth,synth_btcs[i])[0], method='bounded', bounds=range_t_scale) for i in range(n_synth)]
    t_scales = [out.x for out in outs_RMSE_t_scale]
    RMSDs = [out.fun for out in outs_RMSE_t_scale]
    btcs_interp_scaled = [RMSE_t_scale(i_btc,t_scales[i],ts_synth,synth_btcs[i])[1] for i in range(n_synth)]
    a_facs_est = np.array([RMSE_t_scale(i_btc,t_scales[i],ts_synth,synth_btcs[i])[2] for i in range(n_synth)])
    RMSDs_NNI[i_btc] = np.minimum.accumulate(RMSDs)
    NNI_indices = np.array([np.argmin(RMSDs[:i+1]) for i in range(n_synth)])
    vs_est = xs_m[i_btc]/(t_scales/ts_peak_synth)
    est_params_NNI[i_btc] = np.concatenate((a_facs_est[NNI_indices][:,None],vs_est[NNI_indices][:,None],rand_params_synth[NNI_indices]),axis = 1)
    print(i_btc,np.min(RMSDs))
```

```python
def RMSE_t_scale_Z(i_btc,t_scale,Zs_i,t_peak_synth_interp,btc_mean,phis):
    btc_mean_resampled = np.interp(ts_m_0[i_btc],ts_synth*t_scale/t_peak_synth_interp,btc_mean)
    phis_resampled = np.array([np.interp(ts_m_0[i_btc],ts_synth*t_scale/t_peak_synth_interp,phi) for phi in phis.T]).T
    btc_interp =  (btc_mean_resampled+Zs_i@(phis_resampled).T)
    a_fac = np.trapezoid(btc_interp*btcs_m_0[i_btc],x = ts_m_0[i_btc])/np.trapezoid(btc_interp**2,x = ts_m_0[i_btc])
    btc_interp_scaled = btc_interp*a_fac
    RMSD = np.sqrt(np.trapezoid((btcs_m_0[i_btc]-btc_interp_scaled)**2,x = ts_m_0[i_btc])/np.trapezoid((btcs_m_0[i_btc])**2,x = ts_m_0[i_btc]))
    return RMSD,btc_interp_scaled
```

```python
%%time


RMSDs_linear = np.zeros((n_btcs,len(ns_synth_interm_linear)))
est_params_linear = np.zeros((n_btcs,len(ns_synth_interm_linear),5))

for i_btc in range(n_btcs):
    if len(btcs_m[i_btc])>N_terms:
        weight_reg = 0.0001
    else:
        weight_reg = 0.001
    np.random.seed(global_seed)
    bounds = np.concatenate((np.array([-np.inf,np.log(5000)])[None,:],
                             np.array([-np.inf,np.inf])*np.ones(2)[:,None]))
    w = np.trapezoid(np.eye(len(ts_m_0[i_btc])),x= ts_m_0[i_btc])

    range_t_scale_rel_val = 0.8
    index_max_first,index_max_last = np.where(btcs_m_0[i_btc]>range_t_scale_rel_val*np.max(btcs_m_0[i_btc]))[0][[0,-1]]
    def interp_rel_val(index):
        return (ts_m_0[i_btc][index-1]
                +(ts_m_0[i_btc][index]-ts_m_0[i_btc][index-1])
                /(btcs_m_0[i_btc][index]/np.max(btcs_m_0[i_btc])-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc]))
                *(range_t_scale_rel_val-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc])))
    range_t_scale = [interp_rel_val(index_max_first),interp_rel_val(index_max_last+1)]

    for j,n_synth_interm_linear in enumerate(ns_synth_interm_linear):

        def Zs_interpolate(pars,tri,tree,Zs):
            ND_interp_vertices,ND_interp_weights = linear_ND_interp(tri,tree,pars)
            return ND_interp_weights@Zs[ND_interp_vertices],ND_interp_weights@ts_peak_synth[:n_synth_interm_linear][ND_interp_vertices]
        
        triangulation = Delaunay(np.log(rand_params_synth[:n_synth_interm_linear]))
        tree = cKDTree(np.log(rand_params_synth[:n_synth_interm_linear]))
        
        RMSD_denom = np.sum(w*(btcs_m_0[i_btc])**2)
        def loss(par):
            Zs_i,t_peak_synth_interp = Zs_interpolate(par,triangulation,tree,Zs_all[j,:n_synth_interm_linear,:min(N,n_synth_interm_linear)])
            return sp.optimize.minimize_scalar(lambda x: RMSE_t_scale_Z(i_btc,x,Zs_i,t_peak_synth_interp,
                    btc_mean_all[j],phis_all[j,:,:min(N,n_synth_interm_linear)])[0], method='bounded', bounds=range_t_scale).fun + weight_reg * np.sum(Zs_i**2/lambdas_all[j,:min(N,n_synth_interm_linear)])
        results = [sp.optimize.minimize(loss,x0 = np.log(est_params_NNI[i_btc,n_synth_interm_linear-1,2:]),method='Nelder-Mead',bounds = bounds)]
        RMSDs_interp = [result.fun for result in results]
        index = np.argmin(RMSDs_interp)
        est_linear = np.exp(results[index].x)
        btc_linear = forward.laplace.ADEMF_1D(ts_synth,[1.0],1.0,est_linear[0],est_linear[1]/est_linear[2],[est_linear[2],est_linear[2]],'first order','infinite')[:,-1]
        t_scale_linear = sp.optimize.minimize_scalar(lambda x: RMSE_t_scale(i_btc,x,ts_synth,btc_linear)[0], method='bounded', bounds=range_t_scale).x
        RMSD,btc_interp_scaled,a_fac = RMSE_t_scale(i_btc,t_scale_linear,ts_synth,btc_linear)
        v_est = xs_m[i_btc]/(t_scale_linear/ts_synth[np.argmax(btc_linear)])
        est_params_linear[i_btc,j] = [a_fac,v_est,*est_linear]
        RMSDs_linear[i_btc,j] = min(RMSDs_NNI[i_btc,n_synth_interm_linear-1],RMSD)
    print(i_btc,RMSDs_NNI[i_btc,-1],RMSD,RMSDs_interp[index])
```

```python
%%time

np.random.seed(42)

n_pop = 50
n_iter = 2000
log_pars_chain = np.zeros((len(i_btcs_rand),n_iter+1,n_pop,3))
a_fac_chain = np.zeros((len(i_btcs_rand),n_iter+1,n_pop))
t_scale_chain = np.zeros((len(i_btcs_rand),n_iter+1,n_pop))
l_chain = np.zeros((len(i_btcs_rand),n_iter+1,n_pop))
sigma_proc_chain = np.zeros((len(i_btcs_rand),n_iter+1,n_pop))
sigma_noise_chain = np.zeros((len(i_btcs_rand),n_iter+1,n_pop))

log_posts = np.zeros((len(i_btcs_rand),n_iter+1,n_pop))

eps_std = 3e-4
accept_rates = np.zeros((len(i_btcs_rand),n_iter))

beta_l_fac = 0.5
sigma_sigma_proc = 0.005#0.2
sigma_sigma_noise_fac = 0.005
range_t_scale_rel_val = 0.8

min_a_scale = 0.5
max_a_scale = 2.0

d = 8

for kk,i_btc in enumerate(i_btcs_rand):

    def log_prior_hyper(l,sigma_proc,sigma_noise):
        lp_l = sp.stats.invgamma.logpdf(l, 1.0, loc=0, scale=ts_peak[i_btc]*beta_l_fac)
        lp_sigma_proc = sp.stats.halfnorm.logpdf(sigma_proc,loc=0, scale=sigma_sigma_proc)
        lp_sigma_noise = sp.stats.halfnorm.logpdf(sigma_noise,loc=0, scale=np.max(btcs_m[i_btc])*sigma_sigma_noise_fac)
        return lp_l+lp_sigma_proc+lp_sigma_noise
    
    
    index_max_first,index_max_last = np.where(btcs_m_0[i_btc]>range_t_scale_rel_val*np.max(btcs_m_0[i_btc]))[0][[0,-1]]
    def interp_rel_val(index):
        return (ts_m_0[i_btc][index-1]
                +(ts_m_0[i_btc][index]-ts_m_0[i_btc][index-1])
                /(btcs_m_0[i_btc][index]/np.max(btcs_m_0[i_btc])-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc]))
                *(range_t_scale_rel_val-btcs_m_0[i_btc][index-1]/np.max(btcs_m_0[i_btc])))
    range_t_scale = [interp_rel_val(index_max_first),interp_rel_val(index_max_last+1)]
    
    def Zs_interpolate(pars,tri,tree,Zs):
        ND_interp_vertices,ND_interp_weights = linear_ND_interp(tri,tree,pars)
        return ND_interp_weights@Zs[ND_interp_vertices],ND_interp_weights@ts_peak_synth[ND_interp_vertices]
    
    triangulation = Delaunay(np.log(rand_params_synth))
    tree = cKDTree(np.log(rand_params_synth))
    
    def log_prior(a_fac_rel,t_scale,log_pars):
        lp_a_fac = sp.stats.uniform.logpdf(a_fac_rel, loc=min_a_scale, scale=max_a_scale-min_a_scale)
        lp_t_v = sp.stats.uniform.logpdf(t_scale, loc=range_t_scale[0], scale=(range_t_scale[1]-range_t_scale[0]))
        lp_pars = sp.stats.multivariate_normal.logpdf(log_pars, mean=mean_log_params[1:], cov=cov_log_params[1:][:,1:])
        lp_pars[np.exp(log_pars[:,0])>5000] = -np.inf
        return lp_a_fac+lp_t_v+lp_pars
    
    def log_likelihood(a_fac_rel,t_scale,log_pars,l,sigma_proc,sigma_noise):
        Zs_i,t_peak_synth = Zs_interpolate(log_pars,triangulation,tree,Zs)
        btc_mean_resampled = np.interp(ts_m[i_btc],ts_synth*t_scale/t_peak_synth,btc_mean)
        phis_resampled = np.array([np.interp(ts_m[i_btc],ts_synth*t_scale/t_peak_synth,phi) for phi in phis.T]).T
        btc_resampled =  (btc_mean_resampled+Zs_i@(phis_resampled).T)
        a_fac_ref = np.trapezoid(btc_resampled*btcs_m[i_btc],x = ts_m[i_btc])/np.trapezoid(btc_resampled**2,x = ts_m[i_btc])
        btc_resampled_scaled = btc_resampled*a_fac_ref*a_fac_rel
        mask = btc_resampled_scaled>np.max(btc_resampled_scaled)*1e-3
        model = btc_resampled_scaled[mask]
        meas = btcs_m[i_btc][mask]
        t = ts_m[i_btc][mask]
        N_m = len(meas)
        W = np.diag(np.trapezoid(np.eye(N_m),x = t))
        dits = t-t[:,None]
        K = sigma_proc**2*np.exp(-(dits)**2/l**2)
        Sigma = (model*np.eye(N_m))@K@(model*np.eye(N_m))+sigma_noise**2*np.eye(N_m)
        Sigma += np.eye(N_m) * (np.max(btcs_m_0[i_btc])*0.0005)**2
        term1 = -1/2*(W@(meas-model)).T@np.linalg.solve(W@Sigma@W,W@(meas-model))
        term2 = -1/2*np.log(np.linalg.det(W@Sigma@W))
        term3 = -N_m/2*np.log(2*np.pi)
        if term1+term2+term3==np.inf:
            print(term1,term2,term3)
        return term1+term2+term3

    
    # rand_a_fac_rel_pop = np.random.uniform(min_a_scale,max_a_scale,size = n_pop)
    # rand_t_scale_pop = np.random.uniform(*range_t_scale,size = n_pop)
    rand_a_fac_rel_pop = np.ones(n_pop)
    rand_t_scale_pop = np.ones(n_pop)*ts_peak[i_btc]
    
    rand_log_params_pop = np.random.multivariate_normal(np.log(est_params_linear[i_btc,-1][2:]), cov_log_params[1:][:,1:]/4, size=n_pop*3)
    rand_log_params_pop = rand_log_params_pop[np.exp(rand_log_params_pop[:,0])<5000]
    rand_log_params_pop = rand_log_params_pop[:n_pop]
    
    
    rand_l_pop = sp.stats.invgamma.rvs(3.0, loc=0, scale=ts_peak[i_btc]*beta_l_fac, size=n_pop)
    rand_sigma_proc_pop = sp.stats.halfnorm.rvs(loc=0, scale=sigma_sigma_proc, size=n_pop)
    rand_sigma_noise_pop = sp.stats.halfnorm.rvs(loc=0, scale=np.max(btcs_m[i_btc])*sigma_sigma_noise_fac, size=n_pop)
    
    scales = [max_a_scale-min_a_scale,range_t_scale[1]-range_t_scale[0],cov_log_params[1:][:,1:],ts_peak[i_btc]*beta_l_fac,sigma_sigma_proc,np.max(btcs_m[i_btc])*sigma_sigma_noise_fac]
    
    population = [rand_a_fac_rel_pop,rand_t_scale_pop,rand_log_params_pop,rand_l_pop,rand_sigma_proc_pop,rand_sigma_noise_pop]
    
    lp_hyper = log_prior_hyper(*population[3:])
    lp = log_prior(*population[:3])
    ll = np.array([log_likelihood(*[pop[i] for pop in population]) for i in range(n_pop)])
    lpost = lp_hyper+lp+ll
    
    a_fac_chain[kk][0] = population[0]
    t_scale_chain[kk][0] = population[1]
    log_pars_chain[kk][0] = population[2]
    
    l_chain[kk][0] = population[3]
    sigma_proc_chain[kk][0] = population[4]
    sigma_noise_chain[kk][0] = population[5]

    log_posts[kk,0] = lpost
    for ii in range(n_iter):

        if ii%10 ==0:
            gamma = 1.0
        else:
            gamma = 0.8 / np.sqrt(d)
    
        indices = np.arange(n_pop)*np.ones(n_pop)[:,None]
        i, j = np.indices(indices.shape)
        indices = indices[i != j].reshape((n_pop,n_pop-1))
        
        
        r,s = np.array([np.random.choice(indices[i], 2, replace=False) for i in range(n_pop)]).astype(int).T
        
        diff = [pop[r] - pop[s] for pop in population]
    
        epsilons = [np.random.multivariate_normal(np.zeros(len(scale)),scale*eps_std,size = n_pop) if (type(scale)==np.ndarray)
                    else np.random.normal(0.0,scale*eps_std,size = n_pop) for scale in scales]
        
        population_prop = [population[i] + gamma * diff[i] + epsilons[i] for i in range(len(population))]
    
        lp_hyper = log_prior_hyper(*population_prop[3:])
        lp = log_prior(*population_prop[:3])
        ll = np.array([log_likelihood(*[pop[i] for pop in population_prop]) for i in range(n_pop)])
        lpost_prop = lp_hyper+lp+ll
    
        mask_accept =  np.log(np.random.rand(n_pop)) < (lpost_prop - lpost)
        lpost[mask_accept] = lpost_prop[mask_accept] 
       
        accept_rates[kk,ii]=np.mean(mask_accept)

        
        
        for i in range(len(population)):
            population[i][mask_accept] = population_prop[i][mask_accept]
        log_posts[kk,ii+1] = lpost

        if ii%25 == 0:
            print(ii,np.nanmean(lpost[lpost!=-np.inf]))

        if ii%100 == 99:
            mean_chain_val = np.mean(log_posts[kk,ii-99:ii+1],axis = 0)
            q1,q3 = np.percentile(mean_chain_val,[25,75])
            iqr = q3-q1
            mask_outlier = mean_chain_val<q1-5.0*iqr
            print(np.mean(mask_outlier))
            for i in range(len(population)):
                population[i][mask_outlier] = population_prop[i][np.argmax(mean_chain_val)]
            
    
        a_fac_chain[kk][ii+1] = population[0]
        t_scale_chain[kk][ii+1] = population[1]
        log_pars_chain[kk][ii+1] = population[2]
        l_chain[kk][ii+1] = population[3]
        sigma_proc_chain[kk][ii+1] = population[4]
        sigma_noise_chain[kk][ii+1] = population[5]
        
    print(i_btc,np.var(log_pars_chain[kk][-1],axis = 0,ddof = 1)/cov_log_params[np.arange(3)+1,np.arange(3)+1],np.var(log_pars_chain[kk][-1],axis = 0,ddof = 1)**(1/2))
    
```

```python
def gelman_rubin(chains):
    m, n, d = chains.shape
    
    # Mean per chain (m, d)
    chain_means = np.mean(chains, axis=1)
    
    # Global mean (d,)
    global_mean = np.mean(chain_means, axis=0)
    
    # Between-chain variance (d,)
    B = n * np.var(chain_means, axis=0, ddof=1)
    
    # Within-chain variance (d,)
    W = np.mean(np.var(chains, axis=1, ddof=1), axis=0)
    
    # Variance estimate
    V_hat = (n - 1)/n * W + B/n
    
    # R-hat
    R_hat = np.sqrt(V_hat / W)
    
    return R_hat
```

```python
%%time

def log_likes_all(bayes_stats):

    length, sigma_proc, sigma_noise = np.exp(bayes_stats)
    

    log_likes = np.zeros((len(all_rand_params_UQ), len_a_facs))

    for i in range(len(all_rand_params_UQ)):

        n = N_m_all[i]

        K = sigma_proc**2 * np.exp(-(dists_all[i])**2 / length**2)

        a_facs = a_facs_all[i]

        Sigma_1 = (model_all[i]*np.eye(n)) @ K @ (model_all[i]*np.eye(n))
        Sigma_2 = sigma_noise**2 * np.eye(n)

        # residuals
        F = (W_all[i] @ (meas_all[i] - model_all[i]*a_facs[:,None]).T).T

        # covariance matrices
        Sigma = Sigma_1*a_facs[:,None,None]**2 + Sigma_2

        Sigma_scaled = np.array([
            W_all[i] @ S @ W_all[i] for S in Sigma
        ])

        term1 = np.zeros(len_a_facs)
        term2 = np.zeros(len_a_facs)

        for j in range(len_a_facs):

            term1[j] = -0.5 * F[j] @ np.linalg.solve(Sigma_scaled[j], F[j])
            term2[j] = -0.5 * np.log(np.linalg.det(Sigma_scaled[j]))

        term3 = -n/2*np.log(2*np.pi)

        log_likes[i] = term1 + term2 + term3

    return log_likes

def log_likes_int(bayes_stats):
    log_likes = log_likes_all(bayes_stats)
    m = np.max(log_likes)
    log_sum = m + np.log(np.sum(np.exp(log_likes-m)))
    print(log_sum)
    return -log_sum
bayes_stats_0 = np.log([ts_peak[i_btc]*0.05,0.2,0.05*np.max(btcs_m_0[i_btc])])
opt_log_likes_int = sp.optimize.minimize(log_likes_int,x0=bayes_stats_0,method = 'Nelder-Mead', options={'maxfev': 50})
```

## Uncertainty quantification

```python
# %%time
import numpy as np
from scipy.stats import multivariate_normal
n_calls_total = 1000
n_chains = 8
eps_std = 1e-6
sigma = 0.05

est_params_DREAM = np.zeros((n_btcs,n_chains,4))
n_iter = n_calls_total // n_chains
histories_DREAM = np.zeros((n_btcs,n_iter,n_chains))

for i_btc in range(1):

    forward_model = lambda theta: forward.laplace.ADEMF_1D(ts_m_0[i_btc][1:],[xs_m[i_btc]],vs_peak[i_btc]*theta[0],min(theta[1],5000),theta[2]/theta[3],[theta[3],theta[3]],'first order','infinite')[:,-1]
    data = btcs_m_0[i_btc][1:]/np.trapezoid(btcs_m_0[i_btc][1:],x=ts_m_0[i_btc][1:])
    sigma_obs = data*sigma
    np.random.seed(42)   

    d = len(mean_log_params)
    gamma = 0.2*2.38 / np.sqrt(2*d)
    
    # iterations determined by call budget

    # ----- Log posterior -----
    def log_prior(phi):
        return multivariate_normal.logpdf(
            phi, mean=mean_log_params, cov=cov_log_params
        )

    def log_likelihood(phi):
        theta = np.exp(phi)
        y_model = forward_model(theta)        
        RMSD = utils.errors.RMSD_func(ts_m_0[i_btc][1:],data,y_model) 
        log_like = -0.5 * RMSD
        return log_like, RMSD

    # ----- Initialize chains from prior -----
    chains = np.random.multivariate_normal(
        mean_log_params, cov_log_params, size=n_chains
    )

    log_likes_and_RMSDs = [log_likelihood(phi) for phi in chains]
    
    log_posts = np.array([log_prior(phi)+1/sigma**2*l[0] for phi,l in zip(chains,log_likes_and_RMSDs)])
    RMSDs = np.array([l[0] for l in log_likes_and_RMSDs])

    samples = []
    losses_history = []

    # ----- DREAM sampling -----
    for t in range(n_iter):
        for i in range(n_chains):

            # choose r,s different from i
            idx = list(range(n_chains))
            idx.remove(i)
            r, s = np.random.choice(idx, 2, replace=False)

            diff = chains[r] - chains[s]
            eps = np.random.normal(0, eps_std, size=d)

            phi_prop = chains[i] + gamma * diff + eps
            log_like,RMSD = log_likelihood(phi_prop)
            log_pri = log_prior(phi_prop)
            logp_prop = log_prior(phi_prop) + 1/sigma**2*log_like

            # Metropolis accept
            if np.log(np.random.rand()) < (logp_prop - log_posts[i]):
                chains[i] = phi_prop
                log_posts[i] = logp_prop
                RMSDs[i] = RMSD

        samples.append(chains.copy())
        losses_history.append(RMSDs.copy())
        

    samples_post = np.array(samples)
    est_params_DREAM[i_btc] = np.exp(samples_post)[-1]
    histories_DREAM[i_btc] = np.array(losses_history)
```

```python
%%time
import numpy as np
from scipy.stats import multivariate_normal

n_calls_total = 20000
n_chains = 40
eps_std = 1e-6

est_params_DREAM = np.zeros((n_btcs,n_chains,4))  # 4 params + sigma
n_iter = n_calls_total // n_chains
histories_DREAM = np.zeros((n_btcs,n_iter+1,n_chains))
loglikes_histories_DREAM = np.zeros((n_btcs,n_iter+1,n_chains))
rej_ratios = np.zeros(n_btcs)

d_theta = len(mean_log_params)
d = d_theta + 1   # extra dimension for log_sigma

gamma = 0.15

for i_btc in range(1):
    RMSDs = []
    loglikes = []
    count_rej = 0
    forward_model = lambda theta: forward.laplace.ADEMF_1D(
        ts_m_0[i_btc][1:],
        [xs_m[i_btc]],
        vs_peak[i_btc]*theta[0],
        min(theta[1],5000),
        theta[2]/theta[3],
        [theta[3],theta[3]],
        'first order',
        'infinite'
    )[:,-1]

    data = btcs_m_0[i_btc][1:]
    data = data / np.trapezoid(data, x=ts_m_0[i_btc][1:])

    np.random.seed(42)

    
    
    # ----- Log prior -----
    def log_prior(phi_full):
        phi = phi_full[:d_theta]
        log_sigma = phi_full[-1]

        lp_theta = multivariate_normal.logpdf(
            phi, mean=mean_log_params, cov=cov_log_params
        )

        # weak prior on log_sigma
        lp_sigma = -0.5 * (log_sigma / 2.0)**2  # broad Gaussian prior

        return lp_theta + lp_sigma

    # ----- Log likelihood -----
    def log_likelihood(phi_full):
        phi = phi_full[:d_theta]
        log_sigma = phi_full[-1]
        sigma = np.exp(log_sigma)

        theta = np.exp(phi)

        y_model = forward_model(theta)
        RMSD = utils.errors.RMSD_func(ts_m_0[i_btc][1:],data,y_model) 
        RMSDs.append(RMSD)
        residual = data - y_model
        sigma_i = sigma * data + np.max(data)*1e-3  # relative noise model

        ll = -0.5 * np.sum(
            (residual / sigma_i)**2 +
            np.log(2*np.pi*sigma_i**2)
        )
        loglikes.append(ll)
        return ll

    # ----- Initialize chains -----
    chains_theta = np.random.multivariate_normal(
        mean_log_params, cov_log_params, size=n_chains
    )

    chains_log_sigma = np.random.normal(
        np.log(0.05), 1.0, size=n_chains
    )

    chains = np.hstack([
        chains_theta,
        chains_log_sigma[:,None]
    ])

    log_posts = np.array([
        log_prior(phi) + log_likelihood(phi)
        for phi in chains
    ])

    samples = []
    losses_history = []

    # ----- DREAM sampling -----
    for t in range(n_iter):
        for i in range(n_chains):

            idx = list(range(n_chains))
            idx.remove(i)
            r, s = np.random.choice(idx, 2, replace=False)

            diff = chains[r] - chains[s]
            eps = np.random.normal(0, eps_std, size=d)

            phi_prop = chains[i] + gamma * diff + eps

            logp_prop = log_prior(phi_prop) + log_likelihood(phi_prop)

            if np.log(np.random.rand()) < (logp_prop - log_posts[i]):
                chains[i] = phi_prop
                log_posts[i] = logp_prop
            else:
                count_rej += 1

        samples.append(chains.copy())

    samples_post = np.array(samples)
    histories_DREAM[i_btc] = np.array(RMSDs).reshape((n_iter+1,n_chains))
    loglikes_histories_DREAM[i_btc] = np.array(loglikes).reshape((n_iter+1,n_chains))
    est_params_DREAM[i_btc] = np.exp(samples_post[-1,:,:d_theta])
    rej_ratios[i_btc] = count_rej/(n_iter+1*n_chains)
```

```python
import numpy as np
from scipy.optimize import minimize
from scipy.linalg import cho_factor, cho_solve
import matplotlib.pyplot as plt


def kernel_se(x1, x2, sigma_f, ell):
    x1 = x1[:, None]
    x2 = x2[None, :]
    sqdist = (x1 - x2)**2
    return sigma_f**2 * np.exp(-0.5 * sqdist / ell**2)

def neg_log_likelihood_emp_bayes(params,f_thetas, x, y):

    N = len(f_thetas)
    
    log_sigma_f, log_ell, log_sigma_n = params

    sigma_f = np.exp(log_sigma_f)
    ell = np.exp(log_ell)
    sigma_n = np.exp(log_sigma_n)

    
    
    # covariance
    K = kernel_se(x, x, sigma_f, ell)
    K += sigma_n**2 * np.eye(len(x))

    # Cholesky
    L, lower = cho_factor(K, lower=True)

    residual_norms = np.zeros(len(f_thetas))
    for i,f_theta in enumerate(f_thetas):
        f_theta[np.isnan(f_theta)] = np.interp(x[np.isnan(f_theta)],x[~np.isnan(f_theta)],f_theta[~np.isnan(f_theta)])
        r = y - f_theta
        alpha = cho_solve((L, lower), r)
        residual_norms[i] = r @ alpha

    log_det = 2 * np.sum(np.log(np.diag(L)))

    n = len(x)

    m = np.max(residual_norms)
    log_sum = m + np.log(np.sum(np.exp(residual_norms-m)))
    
    return 0.5 * log_sum - np.log(N) + 0.5 * log_det + 0.5 * n * np.log(2*np.pi)

# -------------------------------------------------
# 4. Generate synthetic data
# -------------------------------------------------

np.random.seed(42)

x = ts_synth[synth_btcs[0]>1e-3*np.max(synth_btcs[0])]
n = len(x)

y_clean = np.log(synth_btcs[0][synth_btcs[0]>1e-3*np.max(synth_btcs[0])])

# true GP noise
K_true = kernel_se(x, x, sigma_f=0.3, ell=0.5)
L = np.linalg.cholesky(K_true + 1e-6*np.eye(n))
noise = L @ np.random.randn(n)

y = y_clean + noise
```

```python
import numpy as np
from scipy.optimize import minimize_scalar

def empirical_bayes_sigma2(S,n):
    N = len(S)
    def neg_log_marginal(log_sigma2):
        sigma2 = np.exp(log_sigma2)

        a = -S / (2 * sigma2)
        m = np.max(a)
        log_sum = m + np.log(np.sum(np.exp(a - m)))

        log_marginal = (
            - (n/2) * np.log(2*np.pi*sigma2)
            + log_sum
            - np.log(N)
        )

        return -log_marginal  # minimize

    result = minimize_scalar(
        neg_log_marginal,
        bounds=(-20, 10),
        method='bounded'
    )

    sigma2_hat = np.exp(result.x)
    return sigma2_hat
```

```python
%%time
from lipo import GlobalOptimizer
params_LIPO_2 = np.zeros((n_btcs,4))
KL_divs_LIPO_2 = np.zeros((n_btcs))
RMSEs_LIPO_2 = np.zeros((n_btcs))
btcs_LIPO_2 = []
lipo_errors_mat_2 = []
params_LIPO_2_steps = []
num_function_calls = 300
for i_btc in range(n_btcs):
    lipo_errors_list = []
    params_step = []
    t,btc,x = ts_m_0[i_btc],btcs_m_0[i_btc],xs_m[i_btc]
    def opt_function(v_eps,Pe,beta_prime, mem_param):
        params_step.append([v_eps,Pe,beta_prime, mem_param])
        if t[0]==0:
            est_bt =  np.concatenate(([0],c_lap_solver(t[1:],[x],vs_peak[i_btc]*v_eps, Pe,beta_prime,[1,mem_param],memory_func, bound_cond)[:,-1]))
        else:
            est_bt =  c_lap_solver(t,[x],vs_peak[i_btc]*v_eps, Pe,beta_prime,[1,mem_param],memory_func, bound_cond)[:,-1]
        if memory_func == 'power law':
            obj = -RMSE_func(t,btc,est_bt)-0.02*(v_eps-1)**2
            lipo_errors_list.append(obj)
            return obj
        else:
            obj = -RMSE_func(t,btc,est_bt) 
            lipo_errors_list.append(obj)
            return obj
    pre_eval_x = dict(v_eps = (best_params[:,0]/vs_peak)[i_btc], Pe = best_params[i_btc,1], beta_prime = best_params[i_btc,2],mem_param = best_params[i_btc,3])
    evaluations = [(pre_eval_x, opt_function(**pre_eval_x))]
    search = GlobalOptimizer(
        opt_function,
        lower_bounds={'v_eps': (best_params[:,0]/vs_peak)[i_btc]*0.98,'Pe': best_params[i_btc,1]/1.3,'beta_prime': best_params[i_btc,2]/1.3, 'mem_param': best_params[i_btc,3]/1.3},
        upper_bounds={'v_eps': (best_params[:,0]/vs_peak)[i_btc]*1.02,'Pe': best_params[i_btc,1]*1.3,'beta_prime': best_params[i_btc,2]*1.3, 'mem_param': best_params[i_btc,3]*1.3} ,
        evaluations=evaluations,
        maximize=True
    )
    search.run(num_function_calls)
    v_eps,Pe,beta_prime, mem_param = list(search.optimum[0].values())
    if t[0]==0:
        est_bt = np.concatenate(([0],c_lap_solver(t[1:],[x],vs_peak[i_btc]*v_eps, Pe,beta_prime,[1,mem_param],memory_func, bound_cond)[:,-1]))
    else:
        est_bt = c_lap_solver(t,[x],vs_peak[i_btc]*v_eps, Pe,beta_prime,[1,mem_param],memory_func, bound_cond)[:,-1]

    params_LIPO_2[i_btc] = np.array([vs_peak[i_btc]*v_eps,Pe,beta_prime, mem_param])
    RMSEs_LIPO_2[i_btc] = RMSE_func(t,btc,est_bt)
    KL_divs_LIPO_2[i_btc] = KL_div_func(t,btc,est_bt)
    btcs_LIPO_2.append(est_bt)
    lipo_errors_mat_2.append(lipo_errors_list)
    params_LIPO_2_steps.append(params_step)
    print(i_btc, params_LIPO_2[i_btc], RMSEs_LIPO_2[i_btc])
```

```python
lipo_errors_mat_corr_2 = -np.array([[max(lipo_errors_list[:i+1]) for i in range(num_function_calls)] for lipo_errors_list in lipo_errors_mat_2])
params_LIPO_2_steps_corr = -np.array([[params_LIPO_2_steps[i_btc][np.argmax(lipo_errors_mat_2[i_btc][:i+1])] for i in range(num_function_calls)] for i_btc in range(n_btcs)])
plt.plot(np.exp(np.mean(np.log(lipo_errors_mat_corr_2),axis = 0)))
```
