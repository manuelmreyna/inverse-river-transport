```python
import numpy as np
import matplotlib.pyplot as plt
import utils.load_btcs
conc = np.concatenate
zer = np.zeros
Nt = 150

n_btcs, ts_m, btcs_m, xs_m, names_list, T_ = utils.load_btcs.data('data/river_solute_tracer_data.csv',complete_zeros = True, min_dimless_res = 1/Nt)
```

```python
names_test,index_trac_test = np.unique([' '.join(name.split(' ')[:2])+(name[name.find('S')+1])*(name[name.find('S')+1]).isalpha() for name in names_list],return_inverse=True)
index_ttl = [np.arange(len(names_list))[index_trac_test==i] for i in set(index_trac_test)]
```

```python
def cv_f(t,btc):
    m1 = np.trapz(btc*t,t)/np.trapz(btc,t)
    m2 = np.trapz(btc*(t-m1)**2,t)/np.trapz(btc,t)
    return m2**(1/2)/m1
def var_f(t,btc):
    m1 = np.trapz(btc*t,t)/np.trapz(btc,t)
    m2 = np.trapz(btc*(t-m1)**2,t)/np.trapz(btc,t)
    return m2
def csk_f(t,btc):
    m1 = np.trapz(btc*t,t)/np.trapz(btc,t)
    m2 = np.trapz(btc*(t-m1)**2,t)/np.trapz(btc,t)
    m3 = np.trapz(btc*(t-m1)**3,t)/np.trapz(btc,t)
    return m3/m2**(3/2)
def kurt_f(t,btc):
    m1 = np.trapz(btc*t,t)/np.trapz(btc,t)
    m2 = np.trapz(btc*(t-m1)**2,t)/np.trapz(btc,t)
    m3 = np.trapz(btc*(t-m1)**3,t)/np.trapz(btc,t)
    m4 = np.trapz(btc*(t-m1)**4,t)/np.trapz(btc,t)
    return m4/m2**(2)
```

```python
ts_m_ttl = [[ts_m[j] for j in i] for i in index_ttl]
btcs_m_ttl = [[btcs_m[j] for j in i] for i in index_ttl]
xs_m_ttl = [[xs_m[j] for j in i] for i in index_ttl]
covs_m_ttl = [[cv_f(ts_m[j],btcs_m[j]) for j in i] for i in index_ttl]
csks_m_ttl = [[csk_f(ts_m[j],btcs_m[j]) for j in i] for i in index_ttl]
kurt_m_ttl = [[kurt_f(ts_m[j],btcs_m[j]) for j in i] for i in index_ttl]
var_m_ttl = [[var_f(ts_m[j],btcs_m[j]) for j in i] for i in index_ttl]
```

```python
import scipy as sp
import matplotlib.colors as mcolors

fig, axes = plt.subplots(figsize=(5, 3.5))

Pes_ADE = np.linspace(0,10000,10000)
plt.scatter([i for j in covs_m_ttl for i in j],[i for j in csks_m_ttl for i in j], s = 5.0, c = [i for j in xs_m_ttl for i in j], norm=mcolors.LogNorm(), label = 'Measured breakthrough curves')
cb = plt.colorbar()
cb.ax.set_ylabel('Sampling distance (km)')

plt.plot((2*Pes_ADE+8)**(1/2)/(Pes_ADE+2),(12*Pes_ADE+64)/(2*Pes_ADE+8)**(3/2), c = 'k', linestyle = 'dashed',label = r'Predicted by the ADE, $n_c = 2$')
plt.plot((2*Pes_ADE)**(1/2)/(Pes_ADE),(12*Pes_ADE)/(2*Pes_ADE)**(3/2), c = 'k', linestyle = 'dotted',label = r'Predicted by the ADE, $n_c = 0$')
plt.xscale('log')
plt.grid()
plt.xlabel(r'$CV = \frac{m_2^{1/2}}{m_1}$')
plt.ylabel(r'$CSK = \frac{m_3}{m_2^{3/2}}$')
Pes_pred = np.array([sp.optimize.root_scalar(lambda Pe: (2*Pe+8)**(1/2)/(Pe+2)-i,x0 =1.0).root for j in covs_m_ttl for i in j])
plt.legend()
```

## Variable dispersion

```python
int_D = lambda t,D_inf,D_0,alpha_D: D_inf*t-alpha_D*(D_inf-D_0)*(1-np.exp(-t/alpha_D))
ADE_an_sol = lambda x,t,v,D_inf,D_0,alpha_D: (4*np.pi*int_D(t,D_inf,D_0,alpha_D))**(-1/2)*np.exp(-(x-v*t)**2/(4*int_D(t,D_inf,D_0,alpha_D)))
```

```python
def norm_cent_moments(t,btc, n):
    m0 = np.trapz(btc,t)
    m1 = np.trapz(btc*t,t)/m0
    return [m1]+[np.trapz(btc*(t-m1)**k,t)/m0 for k in range(2,n+1)]
```

```python
ts = np.linspace(0,500,100001)[1:]
Pes_ADE = 10**np.linspace(-1.0,5,200)
alphas = 10**np.linspace(-1,2,4)
csk_max_exp = np.zeros((len(alphas),len(Pes_ADE)))
cov_max_exp = np.zeros((len(alphas),len(Pes_ADE)))
kurt_max_exp = np.zeros((len(alphas),len(Pes_ADE)))
for i,alpha_D in enumerate(alphas):
    for j,Pe in enumerate(Pes_ADE):
        m1,m2,m3,m4 = norm_cent_moments(ts,ADE_an_sol(1.0,ts,1.0,1/Pe,0.0,alpha_D), 4)
        csk_max_exp[i][j] = (m3/m2**(3/2))
        cov_max_exp[i][j] = (m2**(1/2)/m1)
        kurt_max_exp[i][j] = (m4/m2**(2))
```

```python
import scipy as sp
import matplotlib.colors as mcolors

fig, axes = plt.subplots(figsize=(7.3, 4))

plt.scatter([i for j in covs_m_ttl for i in j],[i for j in csks_m_ttl for i in j], s = 5.0, c = [i for j in xs_m_ttl for i in j], norm=mcolors.LogNorm(), label = 'Meas. BTCs')
cb = plt.colorbar(pad = 0.01)
cb.ax.set_ylabel('Sampling distance (km)')

plt.plot((2*Pes_ADE+8)**(1/2)/(Pes_ADE+2),(12*Pes_ADE+64)/(2*Pes_ADE+8)**(3/2), c = 'k',label = r'Const. coeff., $n_c = 2$')
plt.plot((2*Pes_ADE)**(1/2)/(Pes_ADE),(12*Pes_ADE)/(2*Pes_ADE)**(3/2), c = 'k', linestyle = 'dashed',label = r'Const. coeff., $n_c = 0$')
for i,alpha_D in enumerate(alphas):
    plt.plot(cov_max_exp[i],csk_max_exp[i], label = fr'$D(t)\propto 1-e^{{-x/{alpha_D}}}$')
plt.xscale('log')
plt.grid()
plt.xlabel(r'$CV = \frac{m_2^{1/2}}{m_1}$')
plt.ylabel(r'$CSK = \frac{m_3}{m_2^{3/2}}$')
plt.xlim(0.01,1.3)
plt.ylim(-2,8.5)
Pes_pred = np.array([sp.optimize.root_scalar(lambda Pe: (2*Pe+8)**(1/2)/(Pe+2)-i,x0 =1.0).root for j in covs_m_ttl for i in j])
plt.legend()
```

```python
from scipy.integrate import cumulative_trapezoid
from scipy.interpolate import interp1d
skw_x_s = []
t_norm_s = []
for j in range(len(ts_m_ttl)):
    t_interp = np.linspace(0,max([t[-1] for t in ts_m_ttl[j]]),2000)
    U_interp = np.linspace(0,1,10000)
    btcs_cum_inv = np.zeros((len(btcs_m_ttl[j])+1, len(U_interp)))
    for i in range(len(btcs_m_ttl[j])):
        ts,btc,x = ts_m_ttl[j][i],btcs_m_ttl[j][i],xs_m_ttl[j][i]
        btcs_cum_inv[i+1] = np.interp(U_interp,cumulative_trapezoid(btc, ts, initial=0)/cumulative_trapezoid(btc, ts, initial=0)[-1],ts)
    x_interp = np.linspace(0,xs_m_ttl[j][-1],1000)
    btcs_cum_inv_interp = interp1d(np.concatenate(([0.0],xs_m_ttl[j])),btcs_cum_inv,axis = 0)(x_interp)
    btcs_cum_interp = np.array([interp1d(btc_cum_inv_interp,U_interp, bounds_error = False, fill_value = (0.0,1.0))(t_interp) for btc_cum_inv_interp in btcs_cum_inv_interp])
    btcs_interp = np.gradient(btcs_cum_interp,t_interp, axis = 1).T
    skw_x_s_ = []
    for k in range(np.where(~(btcs_interp[:,-1]<10**-4*np.max(btcs_interp[:,-1])))[0][0]):
        m1_x,m2_x,m3_x = norm_cent_moments(x_interp,btcs_interp[k], 3)
        skw_x_s_.append(m3_x/m2_x**(3/2))
    skw_x_s.append(skw_x_s_)
    t_norm_s.append(t_interp[:np.where(~(btcs_interp[:,-1]<10**-4*np.max(btcs_interp[:,-1])))[0][0]]/(np.trapz(btc*ts,ts)/np.trapz(btc,ts)))
```

```python
import numpy as np
from scipy.interpolate import interp1d

# Define common grid
t_common = np.linspace(0, 0.9, 200)  # 200 points between 0 and 0.8
interpolated = []

for j in range(len(skw_x_s)):
    t = t_norm_s[j]
    y = skw_x_s[j]

    # Only interpolate if curve covers part of [0, 0.8]
    if len(t)>0:
        if np.min(t) <= 0.9 and np.max(t) >= 0.0:
            # Restrict interpolation domain
            t_min = max(np.min(t), 0.0)
            t_max = min(np.max(t), 0.9)
    
            mask = (t_common >= t_min) & (t_common <= t_max)
    
            f = interp1d(t, y, bounds_error=False, fill_value=np.nan)
            y_interp = f(t_common)
    
            # mask out values beyond curve support
            y_interp[~mask] = np.nan
            interpolated.append(y_interp)

# Convert to array
interpolated = np.array(interpolated)

# Median
median_vals = np.nanmedian(interpolated, axis=0)

# Percentiles
p10 = np.nanpercentile(interpolated, 10, axis=0)
p90 = np.nanpercentile(interpolated, 90, axis=0)

fig, ax = plt.subplots(figsize=(5, 4))

ax.plot(t_common, median_vals, c='k', lw=2, label='Median')
ax.plot(t_common, p10, c='r', ls='--', lw=1, label='10th percentile')
ax.plot(t_common, p90, c='r', ls='--', lw=1, label='90th percentile')

# Optionally shade between 10th–90th
ax.fill_between(t_common, p10, p90, color='r', alpha=0.2)

# plt.ylim(-2, 2)
ax.grid()
ax.set_xlabel(r'$\frac{t}{m_1^*}$')
ax.set_ylabel(r'$\frac{m_{3,x}}{m_{2,x}^{3/2}}$')
ax.legend()
```

```python
def invlap(F, t,args, alpha=0, tol=1e-10,M = 400):
    """
    F      = laplace-space function (string refering to an m-file),must
             have form F(s, args), where s is the Laplace parameter 
             and return column vector as result
    t      = column vector of times for which real-space function values
             are sought
    args   = other parameters of F
    alpha  = largest pole of F (default zero)
    tol    = numerical tolerance of approaching pole (default 1e-10)
    f      = vector of real-space values f(t)   
    """
    f = []
    # Split up t vector in pieces of the same order of magnitude, invert one piece
    # at a time. Simultaneous inversion for times covering several orders of
    # magnitudes gives inaccurate results for the small times.
    allt = t
    logallt = np.log10(allt)
    iminlogallt = int(np.floor(np.min(logallt)))
    imaxlogallt = int(np.ceil(np.max(logallt)))

    for ilogt in range(iminlogallt, imaxlogallt + 1):
        t = allt[(logallt >= ilogt) & (logallt < (ilogt + 1))]
        if t.size > 0:  # Maybe no elements in that magnitude
            T = np.max(t) * 2
            gamma = alpha - np.log(tol) / (2 * T)
            # NOTE: The correction alpha -> alpha-log(tol)/(2*T) is not in de Hoog's
            #   paper, but in Mathematica's Mathsource (NLapInv.m) implementation of
            #   inverse transforms
            nt = len(t)
            run = np.arange(2 * M + 1)  # so there are 2M+1 terms in Fourier series expansion

            # Find F argument, call F with it, get 'a' coefficients in power series
            s = gamma + 1j * np.pi * run / T
            a = F(s, *args)
            # a = a.flatten()
            a[0] = a[0] / 2  # zero term is halved

            # Build up e and q tables. Superscript is now row index, subscript column
            # CAREFUL: paper uses null index, so all indices are shifted by 1 here
            e = zer((2 * M + 1, M + 1)+a.shape[1:], dtype=complex)
            q = zer((2 * M, M + 1)+a.shape[1:], dtype=complex)  # Column 0 (here: 1) does not exist
            e[:, 0] = zer((2 * M + 1,)+a.shape[1:])
            q[:, 1] = a[1:2 * M + 1] / a[:2 * M]
            for r in range(2, M + 2):  # Step through columns (called r...)
                e[:2 * (M - r+1), r-1] = q[1:2 * (M - r+1) + 1, r-1] - q[:2 * (M - r+1), r-1] + e[1:2 * (M - r+1) + 1, r - 2]
                if r < M:  # One column fewer for q
                    rq = r + 1
                    q[:2 * (M - rq) + 2, rq-1] = q[1:2 * (M - rq) + 3, rq - 2] * e[1:2 * (M - rq) + 3, rq - 2] / e[:2 * (M - rq) + 2, rq - 2]

            # Build up d vector (index shift: 1)
            d = zer((2 * M + 1,)+a.shape[1:], dtype=complex)
            d[0] = a[0]
#             d[1::2] = -q[:M, 1:M + 1].diagonal()
#             d[2::2] = -e[:M, 1:M + 1].diagonal()
            d[1::2] = -q[0, 1:M + 1]#.T
            d[2::2] = -e[0, 1:M + 1]#.T

            # Build up A and B vectors (index shift: 2)co
            # - now make into matrices, one row for each time
            A = zer((2 * M + 2, nt)+a.shape[1:], dtype=complex)
            B = zer((2 * M + 2, nt)+a.shape[1:], dtype=complex)
            A[1, :] = d[0] * np.ones((nt,)+a.shape[1:])
            B[:2, :] = np.ones((2, nt)+a.shape[1:])
            target_shape = (len(t),) + a.shape[1:]
            reshape_dims = (len(t),) + (1,) * (len(a.shape[1:]))
            z = np.exp(1j * np.pi * t / T).reshape(reshape_dims) * np.ones(target_shape)  # Row vector
            
            
            
            dz = d[:,None] * z[None,:]

            for n in range(2, 2 * M + 2):
                A[n, :] = A[n - 1, :] + dz[n - 1, :] * A[n - 2, :]
                B[n, :] = B[n - 1, :] + dz[n - 1, :] * B[n - 2, :]

            # Double acceleration
            h2M = 0.5 * (1 + (d[2 * M - 1] - d[2 * M]) * z)
            R2Mz = -h2M * (1 - np.sqrt(1 + d[2 * M] * z / (h2M**2)))
            A[2 * M + 1, :] = A[2 * M, :] + R2Mz * A[2 * M - 1, :]
            B[2 * M + 1, :] = B[2 * M, :] + R2Mz * B[2 * M - 1, :]

            # Inversion, vectorized for times, make result a column vector
            fpiece = (1 / T * (np.exp(gamma * t).reshape(reshape_dims) * np.ones(target_shape)) * np.real(A[2 * M + 1, :] / B[2 * M + 1, :]))
            # print(fpiece.shape())
            f.extend(fpiece)

    return np.array(f)
```

```python
n_tb = 8
zs_ext = np.linspace(-1,1,n_tb+1)
delta_z = 2/n_tb
zs = (zs_ext[1:]+zs_ext[:-1])/2
beta = 30
Cf = 0.01
v = 1-zs**6
vs = lambda x: 1-zs**6+0*x[:,None]
vs_ext = lambda x: 1-zs_ext**6+0*x[:,None]

def V_f(x):
    V = zer((len(x),n_tb,n_tb))
    V[:,np.arange(n_tb),np.arange(n_tb)] = vs(x)
    return V
def Q_f(x):
    u_stars = vs_ext(x)*np.sqrt(Cf)
    Ds = np.where((zs_ext+0*x[:,None])>0,0.1+0.0*(zs_ext+0*x[:,None]),0.0001+0.0*(zs_ext+0*x[:,None]))#+0*0.6*u_stars*1/beta
    Q = zer((len(x),n_tb,n_tb))
    Q[:,np.arange(n_tb-1),np.arange(n_tb-1)+1] = 1/delta_z**2*Ds[:,1:-1]
    Q[:,np.arange(1,n_tb),np.arange(1,n_tb)-1] = 1/delta_z**2*Ds[:,1:-1]
    Q[:,np.arange(n_tb),np.arange(n_tb)] = -1/delta_z**2*(Ds[:,1:]+Ds[:,:-1])
    return Q
def D_f_inv(x):
    u_stars = vs(x)*np.sqrt(Cf)
    Ds = 0.6*u_stars*1/beta
    D_inv = zer((len(x),n_tb,n_tb))
    D_inv[:,np.arange(n_tb),np.arange(n_tb)] = 1/Ds
    return D_inv
A_f = lambda x: conc((conc((zer((len(x),n_tb,n_tb)), np.ones(len(x))[:,None,None]*np.eye(n_tb)[None,:,:]),axis = 1),
                                   conc((D_f_inv(x)@(-Q_f(x)), D_f_inv(x)@V_f(x)),axis = 1)),axis = 2)
B_f = lambda x: conc((conc((zer((len(x),n_tb,n_tb)),zer((len(x),n_tb,n_tb))),axis = 1),
                               conc((D_f_inv(x), zer((len(x),n_tb,n_tb))),axis = 1)),axis = 2)

x_list = np.linspace(0,30000,50)[1:]
m1s = zer((len(x_list),n_tb,n_tb))
m2s = zer((len(x_list),n_tb,n_tb))
m3s = zer((len(x_list),n_tb,n_tb))
U = np.max(v)
btcs = []


    
for x_i,x in enumerate(x_list):
    print(x_i)
    ts = np.linspace(x/U*0.5,x/U*4.0,500)[1:]
    x_s = np.linspace(0,x,400)
    A = A_f(x_s)
    B = B_f(x_s)
    A1 = np.trapz(A,x_s,axis = 0)
    B1 = np.trapz(B,x_s,axis = 0)
    # A2 = np.trapz(np.trapz(np.matmul(A[None,:,:,:],A[:,None,:,:])-np.matmul(A[:,None,:,:],A[None,:,:,:]),x_s,axis = 0),x_s,axis = 0)
    # AB = np.trapz(np.trapz(np.matmul(A[None,:,:,:],B[:,None,:,:])+np.matmul(B[None,:,:,:],A[:,None,:,:])-
    #                        np.matmul(A[:,None,:,:],B[None,:,:,:])-np.matmul(B[:,None,:,:],A[None,:,:,:]),x_s,axis = 0),x_s,axis = 0)
    # B2 = np.trapz(np.trapz(np.matmul(B[None,:,:,:],B[:,None,:,:])-np.matmul(B[:,None,:,:],B[None,:,:,:]),x_s,axis = 0),x_s,axis = 0)    
    def c_lap_space(s):
        result = zer((len(s),n_tb,n_tb))
        # result = zer((len(s)))
        for i in range(len(s)):
            M = (A1+s[i]*B1)#+(A2+s[i]*AB+B2*s[i]**2)
            lambdas, S = np.linalg.eig(M)
            exp_lambdas = np.exp(lambdas)
            exp_lambdas[np.isnan(exp_lambdas)|np.isinf(exp_lambdas)] = 0.0
            exp_M = S@np.diag(exp_lambdas)@np.linalg.inv(S)#sp.linalg.expm(M)#
            result[i] = exp_M[:n_tb,:n_tb]
            # result[i] = np.sum(exp_M@(np.eye(n_tb)[int(pos*n_tb)]))
        return result
    # c_lap_space_vec = np.vectorize(c_lap_space)
    btc = invlap(c_lap_space,ts,[])
    btc[np.isnan(btc)] = 0.0
    btcs.append(btc)
    for i in range(n_tb):
        for j in range(n_tb):
            m1s[x_i,i,j] = np.trapz(btc[:,i,j]*ts,ts)/np.trapz(btc[:,i,j],ts)
            m2s[x_i,i,j] = np.trapz(btc[:,i,j]*(ts-m1s[x_i,i,j])**2,ts)/np.trapz(btc[:,i,j],ts)
            m3s[x_i,i,j] = np.trapz(btc[:,i,j]*(ts-m1s[x_i,i,j])**3,ts)/np.trapz(btc[:,i,j],ts)

COV = m2s[:,2,2]**(1/2)/m1s[:,2,2]
CSK = m3s[:,2,2]/m2s[:,2,2]**(3/2)
```
