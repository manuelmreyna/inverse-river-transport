```python
import csv

with open('data/river_solute_tracer_data.csv', 'r') as file:
    data_reader = csv.reader(file)
    data_mat = [[cell for cell in row] for row in data_reader]
    breakthrough_curves_data = [[[data_mat[i][i_btc*5+j] for i in range(len(data_mat))] for j in range(4)] for i_btc in range(len(data_mat[0])//5)]
```

```python
ts_m = [np.array([float(c) for c in btc[0][33:] if c!='']) for btc in breakthrough_curves_data if len([float(c) for c in btc[3][33:] if c!=''])>0 and btc[2][18] not in ['','0'] and btc[2][26] in ['','0','0.0']]
btcs_m = [np.array([float(c) for c in btc[3][33:] if c!='']) for btc in breakthrough_curves_data if len([float(c) for c in btc[3][33:] if c!=''])>0 and btc[2][18] not in ['','0'] and btc[2][26] in ['','0','0.0']]
xs_m = [btc[2][18] for btc in breakthrough_curves_data if len([float(c) for c in btc[3][33:] if c!=''])>0 and btc[2][18] not in ['','0'] and btc[2][26] in ['','0','0.0']]
xs_m = [float(i) if i!='' else 0.0 for i in xs_m]
background_cs_m = [btc[2][25] for btc in breakthrough_curves_data if len([float(c) for c in btc[3][33:] if c!=''])>0 and btc[2][18] not in ['','0'] and btc[2][26] in ['','0','0.0']]
background_cs_m = [float(i) if i!='' else 0.0 for i in background_cs_m]
names_list = [btc[0][31] for btc in breakthrough_curves_data if len([float(c) for c in btc[3][33:] if c!=''])>0 and btc[2][18] not in ['','0'] and btc[2][26] in ['','0','0.0']]
n_btcs = len(ts_m)
rivers_list = [n.split(' ')[0] for n in names_list]
rivers_full_name_list = [btc[2][1] for btc in breakthrough_curves_data if len([float(c) for c in btc[3][33:] if c!=''])>0 and btc[2][18] not in ['','0'] and btc[2][26] in ['','0','0.0']]
tracers_list = [' '.join(n.split(' ')[:2])+' '+''.join([j  if ('S'+j in n.split(' ')[-1]) else '' for j in list('ABCDEFGHIJKLMNOPQRSTUVWXYZ')]) for n in names_list]
unique_tracers_list = list(set(tracers_list))
_,count_tracers = np.unique([unique_tracers_list.index(i) for i in tracers_list], return_counts = True)
np.random.seed(1)
random_shuffle = np.arange(len(count_tracers))
np.random.shuffle(random_shuffle)
partial_ratio = np.cumsum(count_tracers[random_shuffle])/n_btcs
test_ratio = 0.1
indices_test = [j for j in range(n_btcs) if tracers_list[j] in [unique_tracers_list[i] for i in random_shuffle[:np.argmin(np.abs(test_ratio-partial_ratio))+1]]]
indices_train = np.delete(np.arange(n_btcs), indices_test)
```

```python
for i_btc in range(n_btcs):
    if len(ts_m[i_btc])<len(btcs_m[i_btc]):
        btcs_m[i_btc] = btcs_m[i_btc][:len(ts_m[i_btc])]
T_= int(max([t[-1]/t[np.argmax(btc)] for t,btc in zip(ts_m,btcs_m)]))+1
print(T_)
```

```python
ts_m_0 = []
btcs_m_0 = []
for i_btc in range(n_btcs):
    t,btc = ts_m[i_btc],btcs_m[i_btc]
    sorted_indices = np.argsort(t)
    t = t[sorted_indices]
    btc = btc[sorted_indices]
    t_unique,indices_unique = np.unique(t, return_index=True)
    if len(t_unique)!=len(t):
        btc_unique = np.zeros(len(indices_unique))
        for i in range(len(indices_unique)):
            if i<len(indices_unique)-1:
                btc_unique[i] = np.mean(btc[indices_unique[i]:indices_unique[i+1]])
            else:
                btc_unique[i] = np.mean(btc[indices_unique[i]:])
        t = t_unique
        btc = btc_unique
    deltat = np.max(t[1:]-t[:-1])
    t_ext = np.concatenate((np.arange(t[0],0,-deltat)[1:][::-1],t,np.arange(t[-1],T_*t[np.argmax(btc)],deltat)[1:]))
    btc_ext = np.interp(t_ext,t,btc-background_cs_m[i_btc],left = 0.0,right = 0.0)
    btc_ext = btc_ext
    ts_m_0.append(t_ext)
    btcs_m_0.append(btc_ext)
```

```python
def calculate_errors(params_est):
    RMSEs = np.zeros(n_btcs)
    KL_divs = np.zeros(n_btcs)
    for i_btc in range(n_btcs):
        t,btc,x, p_est = ts_m_0[i_btc],btcs_m_0[i_btc],xs_m[i_btc], params_est[i_btc]
        sol = np.concatenate(([0.0],forward.laplace.ADEMF_1D(t[1:],[x],p_est[0],p_est[1],p_est[2],[1,p_est[3]], memory_func, bound_cond)[:,-1]))
        RMSEs[i_btc] = (RMSE_func(t,btc,sol))
        KL_divs[i_btc] = (KL_div_func(t,btc,sol))
    return RMSEs,KL_divs
```

```python
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b', '#e377c2', '#7f7f7f', '#bcbd22', '#17becf']
```

```python
if memory_func == 'first order':
    # bounds = (np.array([20.0,0.1,1.0]),np.array([40000.0,20.0,100.0]))
    bounds = (np.array([20.0,0.0,0.0]),np.array([20000.0,np.inf,np.inf]))
if memory_func == 'power law':
    # bounds = (np.array([20.0,0.003,0.1]),np.array([40000.0,1.0,1.0]))
    bounds = (np.array([20.0,0.01,0.1]),np.array([20000.0,1.0,1.0]))
```

```python
vs_peak = np.array([x/ts_meas[np.argmax(measured_bt)] for ts_meas,measured_bt,x in zip(ts_m_0,btcs_m_0,xs_m)])
```

```python
%%time
def estimate_laplace(i_btc):
    t,btc,x, v_ref = ts_m_0[i_btc],btcs_m_0[i_btc],xs_m[i_btc], vs_peak[i_btc]
    t_pos = t[btc>0]
    btc_pos = btc[btc>0]
    v_prior = x/t[np.argmax(btc)]
    # params0 = np.array([v_prior,np.sqrt(bounds[0][0]*bounds[1][0]),np.sqrt(bounds[0][1]*bounds[1][1]),np.sqrt(bounds[0][2]*bounds[1][2])])
    params0 = np.array([v_prior,np.sqrt(bounds[0][0]*bounds[1][0]),0.1,0.3])
    bounds_i = (np.concatenate(([v_prior*0.9],bounds[0])),np.concatenate(([v_prior*1.5],bounds[1])))
    params = estimators.laplace.ADEMF_1D(t_pos,btc_pos,x,'first order','infinite',params0,bounds_i, 0.06, v_ref)
    if t[0]==0:
        btc_rec = np.concatenate(([0.0],forward.laplace.ADEMF_1D(t[1:],[x],params[0],params[1],params[2],[1,params[3]],'first order','infinite')[:,-1]))
    else:
        btc_rec = forward.laplace.ADEMF_1D(t,[x],params[0],params[1],params[2],[1,params[3]],'first order','infinite')[:,-1]
    RMSE = RMSE_func(t,btc,btc_rec)
    KL_div = KL_div_func(t,btc,btc_rec)
    return params,RMSE,KL_div,btc_rec
output_estimate_laplace = [estimate_laplace(i_btc) for i_btc in range(n_btcs)]
params_laplace = np.array([out[0] for out in output_estimate_laplace])
RMSEs_laplace = np.array([out[1] for out in output_estimate_laplace])
KL_divs_laplace = np.array([out[2] for out in output_estimate_laplace])
btcs_rec_laplace = [out[3] for out in output_estimate_laplace]
```

```python
%%time
if memory_func == 'first order':
    def estimate_moments(i_btc):
        t,btc,x = ts_m_0[i_btc],btcs_m_0[i_btc],xs_m[i_btc]
        v_prior = x/t[np.argmax(btc)]
        # params0 =  np.array([v_prior,np.sqrt(bounds[0][0]*bounds[1][0]),np.sqrt(bounds[0][1]*bounds[1][1]),np.sqrt(bounds[0][2]*bounds[1][2])])
        params0 =  np.array([v_prior,np.sqrt(bounds[0][0]*bounds[1][0]),1.0,1.0])
        bounds_i = (np.concatenate(([v_prior*0.8],bounds[0])),np.concatenate(([v_prior*1.5],bounds[1])))
        params = estimators.moments.TSM(t,btc,x,'infinite',params0,bounds_i)
        if t[0]==0:
            btc_rec = np.concatenate(([0.0],forward.laplace.ADEMF_1D(t[1:],[x],params[0],params[1],params[2],[1,params[3]],'first order','infinite')[:,-1]))
        else:
            btc_rec = forward.laplace.ADEMF_1D(t,[x],params[0],params[1],params[2],[1,params[3]],'first order','infinite')[:,-1]
        RMSE = RMSE_func(t,btc,btc_rec)
        KL_div = KL_div_func(t,btc,btc_rec)
        return params,RMSE,KL_div,btc_rec
    output_estimate_moments = [estimate_moments(i_btc) for i_btc in range(n_btcs)]
    params_moments = np.array([out[0] for out in output_estimate_moments])
    RMSEs_moments = np.array([out[1] for out in output_estimate_moments])
    KL_divs_moments = np.array([out[2] for out in output_estimate_moments])
    btcs_rec_moments = [out[3] for out in output_estimate_moments]
```

### Optimize regularization parameter

```python
%%time
n_vs = 11
epsilons = np.linspace(0.95,1.2,n_vs)
reg_Zs = 10**np.linspace(-7,-3,9)
test_errors = np.zeros((len(reg_Zs),n_btcs,n_vs))
for i_reg in range(len(reg_Zs)):
    reg_Z = reg_Zs[i_reg]
    for i_btc in range(n_btcs):
        t, btc, x = ts_m_0[i_btc],btcs_m_0[i_btc],xs_m[i_btc]
        btc = btc/(np.sum(btc)*np.sqrt(len(btc)))
        v_ests = x/t[np.argmax(btc)]*epsilons
        for i in range(n_vs):
            v_est = v_ests[i]
            btcs_mean_interp = np.interp(t/x*v_est, t_, btcs_mean, left=0.0, right=0.0)
            scaling_factor = 1/(np.sum(btcs_mean_interp)*np.sqrt(len(btcs_mean_interp)))
            btcs_mean_interp = btcs_mean_interp*scaling_factor
            phis_interp  = np.array([np.interp(t/x*v_est, t_, phi, left=0.0, right=0.0) for phi in phis[:n_lmbds]])*scaling_factor
            ratio_drop = 0.2
            n_repeats = 5
            index_not_drop = [np.sort(np.random.choice(np.arange(len(btc)),size = int(len(btc)*(1-ratio_drop)),replace=False)) for j in range(n_repeats)]
            index_drop = [np.delete(np.arange(len(btc)),index_not_drop[j]) for j in range(n_repeats)]
            vec_bs = [np.concatenate((btc[index_not_drop[j]]-btcs_mean_interp[index_not_drop[j]],np.zeros(n_lmbds))) for j in range(n_repeats)]
            matrix_Cs = [np.concatenate((phis_interp[:,index_not_drop[j]].T,reg_Z*np.diag(1/np.sqrt(lambdas[:n_lmbds])))) for j in range(n_repeats)]
            Z_js =  [sp.linalg.lstsq(matrix_Cs[j],vec_bs[j])[0] for j in range(n_repeats)]
            rec_btcs = [btcs_mean_interp[index_drop[j]]+phis_interp[:,index_drop[j]].T@Z_js[j] for j in range(n_repeats)]
            test_error = np.sum([np.sum((btc[index_drop[j]]-rec_btcs[j])**2) for j in range(n_repeats)])
            test_errors[i_reg,i_btc,i] = test_error
    print(reg_Z,np.exp(np.mean(np.log(test_errors[i_reg]))))
```

```python
plt.loglog(reg_Zs,np.exp(np.mean(np.mean(np.log(test_errors),axis = -1),axis =-1)))
reg_Z_opt = reg_Zs[np.argmin(np.exp(np.mean(np.mean(np.log(test_errors),axis = -1),axis =-1)))]
```

## ADE

```python
def ADE(t_,Pe):
    return 1/(np.sqrt(4*np.pi*t_/Pe))*np.exp(-(t_-1)**2/(4*t_/Pe))
```

```python
%%time
n_synth_ADE = 20
Pes_rand_ADE = 10**(np.linspace(1.5,3.3,n_synth_ADE))
btcs_synth_ADE = np.array([ADE(t_,Pe) for Pe in Pes_rand_ADE])
```

```python
def acute_triangle(a,b,c):
    return (a**2<b**2+c**2) and (b**2<a**2+c**2) and (c**2<b**2+a**2)
def simplex(points,dim):
    return list(range(dim+1))
def proj(v,u):
    return (v@u)/(u@u)*u
def gs(vectors):
    basis = []
    for vec in vectors:
        u = vec - sum([proj(vec,e) for e in basis])
        basis.append(u/np.linalg.norm(u))
    return basis
def get_projecting_indices(sorted_args,Z,Zs,dim):
    return sorted_args[simplex(np.concatenate(([Z],Zs[sorted_args[:int(dim*len(Zs)**(1/dim))]]),axis = 0), dim)]
def get_projecting_basis(projecting_points):
    return gs(projecting_points[1:]-projecting_points[0])
def get_projected_points(projecting_points,projecting_basis,Z):
    return projecting_points[0]+sum([proj(Z-projecting_points[0],e) for e in projecting_basis])
def get_vertices_basis(projecting_points,projecting_basis):
    return np.array([[(projecting_points[i]-projecting_points[0])@e for e in projecting_basis]  for i in range(len(projecting_points))])
def get_projected_points_basis(projected_points,projecting_points,projecting_basis):
    return np.array([(projected_points-projecting_points[0])@e for e in projecting_basis])
def drop_points_in_simplex(projecting_indices,projecting_points,projecting_basis,projected_points,vertices_basis,projected_points_basis,Z):
    if len(vertices_basis)>1:
        T = (vertices_basis[:-1]-vertices_basis[-1]).T
        bar_coord = np.linalg.solve(T,projected_points_basis-vertices_basis[-1])
        bar_coord = np.concatenate((bar_coord,[1-np.sum(bar_coord)]))
        while np.any(bar_coord<0) and len(projecting_indices)>1:
            projecting_indices = projecting_indices[:-1]
            projecting_points = Zs[projecting_indices,]
            projecting_basis = gs(projecting_points[1:]-projecting_points[0])
            projected_points = projecting_points[0]+sum([proj(Z-projecting_points[0],e) for e in projecting_basis])
            if len(vertices_basis)>=1:
                vertices_basis = np.array([[(projecting_points[i]-projecting_points[0])@e for e in projecting_basis] for i in range(len(projecting_points))])
                projected_points_basis = np.array([(projected_points-projecting_points[0])@e for e in projecting_basis])
                T = (vertices_basis[:-1]-vertices_basis[-1]).T
                bar_coord = np.linalg.solve(T,projected_points_basis-vertices_basis[-1])
                bar_coord = np.concatenate((bar_coord,[1-np.sum(bar_coord)]))
            else:
                bar_coord = np.array([1])
                break
    else:
        bar_coord = np.array([1])
    return bar_coord, projecting_indices,projecting_points,projecting_basis,projected_points,vertices_basis,projected_points_basis
```

```python
def acute_triangle(a,b,c):
    return (a**2<b**2+c**2) and (b**2<a**2+c**2) and (c**2<b**2+a**2)
def simplex(points, dim):
    return list(range(dim+1))
def proj(v,u):
    return (v@u)/(u@u)*u
def gs(vectors):
    basis = []
    for vec in vectors:
        u = vec - sum([proj(vec,e) for e in basis])
        basis.append(u/np.linalg.norm(u))
    return basis
def get_projecting_indices(sorted_args,Z,Zs,dim):
    return sorted_args[simplex(np.concatenate(([Z],Zs[sorted_args[:int(dim*len(Zs)**(1/dim))]]),axis = 0), dim)]
def get_projecting_basis(projecting_points):
    return gs(projecting_points[1:]-projecting_points[0])
def get_projected_points(projecting_points,projecting_basis,Z):
    return projecting_points[0]+sum([proj(Z-projecting_points[0],e) for e in projecting_basis])
def get_vertices_basis(projecting_points,projecting_basis):
    return np.array([[(projecting_points[i]-projecting_points[0])@e for e in projecting_basis]  for i in range(len(projecting_points))])
def get_projected_points_basis(projected_points,projecting_points,projecting_basis):
    return np.array([(projected_points-projecting_points[0])@e for e in projecting_basis])
def drop_points_in_simplex(projecting_indices,projecting_points,projecting_basis,projected_points,vertices_basis,projected_points_basis,Z):
    if len(vertices_basis)>1:
        T = (vertices_basis[:-1]-vertices_basis[-1]).T
        bar_coord = np.linalg.solve(T,projected_points_basis-vertices_basis[-1])
        bar_coord = np.concatenate((bar_coord,[1-np.sum(bar_coord)]))
        while np.any(bar_coord<0) and len(projecting_indices)>1:
            projecting_indices = projecting_indices[:-1]
            projecting_points = Zs[projecting_indices,]
            projecting_basis = gs(projecting_points[1:]-projecting_points[0])
            projected_points = projecting_points[0]+sum([proj(Z-projecting_points[0],e) for e in projecting_basis])
            if len(vertices_basis)>=1:
                vertices_basis = np.array([[(projecting_points[i]-projecting_points[0])@e for e in projecting_basis] for i in range(len(projecting_points))])
                projected_points_basis = np.array([(projected_points-projecting_points[0])@e for e in projecting_basis])
                T = (vertices_basis[:-1]-vertices_basis[-1]).T
                bar_coord = np.linalg.solve(T,projected_points_basis-vertices_basis[-1])
                bar_coord = np.concatenate((bar_coord,[1-np.sum(bar_coord)]))
            else:
                bar_coord = np.array([1])
                break
    else:
        bar_coord = np.array([1])
    return bar_coord, projecting_indices,projecting_points,projecting_basis,projected_points,vertices_basis,projected_points_basis
```
