```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from matplotlib.colors import Normalize
```

```python
def angles_to_xy(ss,thetas,x0,y0):
    ss_diff = np.diff(ss)
    thetas_mean = (thetas[1:]+thetas[:-1])/2
    xs = x0+np.cumsum(ss_diff*np.cos(thetas_mean))
    ys = y0+np.cumsum(ss_diff*np.sin(thetas_mean))
    return xs,ys   
# Convert points to segments for LineCollection
def segments_from_xy(x, y):
    points = np.array([x, y]).T
    return np.stack([points[:-1], points[1:]], axis=1)

def offset_curve(x, y, distance):
    """Offset a 2D curve by a given distance."""
    x = np.asarray(x)
    y = np.asarray(y)
    n = len(x)
    if n < 2:
        raise ValueError("Need at least two points to define a curve.")

    # Compute direction vectors for each segment
    dx = np.diff(x)
    dy = np.diff(y)
    lengths = np.hypot(dx, dy)
    dx /= lengths
    dy /= lengths

    # Normals are perpendicular to direction vectors
    nx = -dy
    ny = dx

    # Compute normals at each point by averaging adjacent segment normals
    normals_x = zer(n)
    normals_y = zer(n)

    normals_x[1:-1] = (nx[:-1] + nx[1:]) * 0.5
    normals_y[1:-1] = (ny[:-1] + ny[1:]) * 0.5

    # Normalize averaged normals
    norm = np.hypot(normals_x, normals_y)
    norm[1:-1] = np.where(norm[1:-1] == 0, 1, norm[1:-1])  # Avoid division by zero
    normals_x[1:-1] /= norm[1:-1]
    normals_y[1:-1] /= norm[1:-1]

    # For endpoints, use the segment normal
    normals_x[0] = nx[0]
    normals_y[0] = ny[0]
    normals_x[-1] = nx[-1]
    normals_y[-1] = ny[-1]

    # Apply the offset
    x_offset = x + distance * normals_x
    y_offset = y + distance * normals_y

    return x_offset, y_offset
```

```python
kinoshita_theta = lambda s,awl,theta0,Js,Jf:theta0*np.sin(2*np.pi*s/awl)+theta0**3*(Js*np.cos(6*np.pi*s/awl)-Jf*np.sin(6*np.pi*s/awl))
kinoshita_curvature = lambda s,awl,theta0,Js,Jf:2*np.pi/awl*(theta0*np.cos(2*np.pi*s/awl)+3*theta0**3*(Js*np.cos(6*np.pi*s/awl+Jf*np.sin(6*np.pi*s/awl))))
kinoshita_vel_grad = lambda s,awl,theta0,Js,Jf,chi,a2,a4: 2*np.pi/awl*(theta0*((-chi+a4*a2/(a2**2+(2*np.pi/awl)**2))*np.cos(2*np.pi*s/awl)+(2*np.pi/awl)*a4/(a2**2+(2*np.pi/awl)**2)*np.sin(2*np.pi*s/awl)) +
                                                                             3*theta0**3*((+Js*chi-Js*a4*a2/(a2**2+(6*np.pi/awl)**2)-Jf*(6*np.pi/awl)*a4/(a2**2+(6*np.pi/awl)**2))*np.sin(6*np.pi*s/awl)+(+Jf*chi-Jf*a4*a2/(a2**2+(6*np.pi/awl)**2)+Js*(6*np.pi/awl)*a4/(a2**2+(6*np.pi/awl)**2))*np.cos(6*np.pi*s/awl)))
# kinoshita_vel_grad = lambda s,awl,theta0,Js,Jf,chi,a2,a4: (-chi+a4/a2)*kinoshita_curvature(s,awl,theta0,Js,Jf)
```

```python
def self_intersection(x,y):
    def int(x1,x2,x3,x4,y1,y2,y3,y4):
        d = (x1-x2)*(y3-y4) - (y1-y2)*(x3-x4)
        if d:
            xs = ((x1*y2-y1*x2)*(x3-x4) - (x1-x2)*(x3*y4-y3*x4)) / d
            ys = ((x1*y2-y1*x2)*(y3-y4) - (y1-y2)*(x3*y4-y3*x4)) / d
            if (xs >= min(x1,x2) and xs <= max(x1,x2) and
                xs >= min(x3,x4) and xs <= max(x3,x4)):
                return xs, ys
    
    x_int, y_int = [], []
    for i in range(len(x)-1):
        for j in range(i-1):
            if xs_ys := int(x[i],x[i+1],x[j],x[j+1],y[i],y[i+1],y[j],y[j+1]):
                x_int.append(xs_ys[0])
                y_int.append(xs_ys[1])
    if len(x_int)>0:
        return True
    else:
        return False
```

```python
for seed in range(0,10):
    np.random.seed(seed)
    beta = np.exp(np.random.normal(loc = 3.01, scale = 0.72))
    awl = np.exp(np.random.normal(loc = 3.7, scale = 0.7))
    theta_0_deg = np.random.uniform(0,120)# 0 to 120 degrees to avoid self-crossings
    theta0 = theta_0_deg/180*np.pi 
    Js = np.random.uniform(-1/20,1/20)
    Jf = np.random.uniform(-1/20,1/20)
    Fr = np.exp(np.random.normal(loc = -2.08, scale = 0.68))
    Cf0 = np.exp(np.random.uniform(-6.9,-3.0))#np.exp(np.random.normal(loc = -3.83, scale = 1.42))
    A = 3.8*(1+beta/6.96*np.exp(-6.96/beta))
    ss = np.linspace(0,awl*2,500)
    thetas = kinoshita_theta(ss,awl,theta0,Js,Jf)
    
    
    xs,ys = angles_to_xy(ss,thetas,0.0,0.0)
    xs_bank_left, ys_bank_left = offset_curve(xs,ys, 1.0)
    xs_bank_right, ys_bank_right = offset_curve(xs,ys, -1.0)
    wl = (xs[-1]-xs[0])/2
    sinuosity = awl/wl #1 to 8 to avoid self-crossings
    chi = sinuosity**(-1/3) # 0.5 to 1.0 to avoid self-crossings
    a2 = 2*Cf0*beta*chi
    a4 = Cf0*beta*chi**2*((1+5*np.sqrt(Cf0))*(A+chi**3*Fr**2)+1)

    vel_grad = kinoshita_vel_grad(ss,awl,theta0,Js,Jf, chi, a2, a4)
    
    vel_bank_left = 1+(vel_grad[1:]+vel_grad[:-1])/2
    vel_bank_right = 1-(vel_grad[1:]+vel_grad[:-1])/2
    
    # Create segments and corresponding LineCollections
    seg1 = segments_from_xy(xs_bank_left, ys_bank_left)
    seg2 = segments_from_xy(xs_bank_right, ys_bank_right)
    
    norm = Normalize(vmin=1+min(vel_grad), vmax=1+max(vel_grad))
    lc1 = LineCollection(seg1, cmap='coolwarm', norm=norm)
    lc1.set_array(vel_bank_left)
    lc1.set_linewidth(2)
    
    lc2 = LineCollection(seg2, cmap='coolwarm', norm=norm)
    lc2.set_array(vel_bank_right)
    lc2.set_linewidth(2)
    
    if not (self_intersection(xs_bank_left, ys_bank_left) or self_intersection(xs_bank_right, ys_bank_right)):
        if np.max(vel_grad)<1.0:
            fig, ax = plt.subplots()
            ax.plot(xs, ys, c='r', linestyle='dashed')
            ax.add_collection(lc1)
            ax.add_collection(lc2)
            
            ax.set_aspect('equal', adjustable='box')
            ax.set_ylim(min(ys)-2.0,max(ys)+2.0)
            ax.set_title('Seed='+str(seed)+r', $\beta$='+str(round(beta,1))+r', $\theta$='+str(round(theta_0_deg,1))+r'$^{\circ}$, $J_s$='+str(round(Js,3))+', $J_s$='+str(round(Jf,3))+', $Fr$='+str(round(Fr,3))+', $C_{f}$='+str(round(Cf0,3)))
            plt.colorbar(lc1, ax=ax, label='Velocity', orientation='horizontal', fraction=0.05, pad=0.1,  aspect=30)  # Optional: show color scale
            plt.show()
            
            print(str(seed)+': No intersect or negative velocity, '+str(theta_0_deg)+', '+str(Js)+', '+str(Jf)+', '+str(sinuosity)+', '+str(chi))
        else:
            print(str(seed)+': Negative velocity, '+str(theta_0_deg)+', '+str(Js)+', '+str(Jf)+', '+str(sinuosity)+', '+str(chi))
    else:
        print(str(seed)+': Intersect, '+str(theta_0_deg)+', '+str(Js)+', '+str(Jf)+', '+str(sinuosity)+', '+str(chi))
    
```

```python
def csk_f(ts,btcs):
    m1 = np.array([np.trapz(btc*t,t)/np.trapz(btc,t) for t,btc in zip(ts,btcs)])
    m2 = np.array([np.trapz(btc*(t-mean)**2,t)/np.trapz(btc,t) for t,btc,mean in zip(ts,btcs,m1)])
    m3 = np.array([np.trapz(btc*(t-mean)**3,t)/np.trapz(btc,t) for t,btc,mean in zip(ts,btcs,m1)])
    return m3/m2**(3/2)
def COV_f(ts,btcs):
    m1 = np.array([np.trapz(btc*t,t)/np.trapz(btc,t) for t,btc in zip(ts,btcs)])
    m2 = np.array([np.trapz(btc*(t-mean)**2,t)/np.trapz(btc,t) for t,btc,mean in zip(ts,btcs,m1)])
    return m2**(1/2)/m1
```

```python
def invlap(F, t,args, alpha=0, tol=1e-8,M = 100):
    """
    F      = laplace-space function (string refering to an m-file),must
             have form F(s, args), where s is the Laplace parameter 
             and return column vector as result
    t      = column vector of times for which real-space function values
             are sought
    args   = other parameters of F
    alpha  = largest pole of F (default zero)
    tol    = numerical tolerance of approaching pole (default 1e-10)
    f      = vector of real-space values f(t)   
    """
    f = []
    # Split up t vector in pieces of the same order of magnitude, invert one piece
    # at a time. Simultaneous inversion for times covering several orders of
    # magnitudes gives inaccurate results for the small times.
    allt = t
    logallt = np.log10(allt)
    iminlogallt = int(np.floor(np.min(logallt)))
    imaxlogallt = int(np.ceil(np.max(logallt)))

    for ilogt in range(iminlogallt, imaxlogallt + 1):
        t = allt[(logallt >= ilogt) & (logallt < (ilogt + 1))]
        if t.size > 0:  # Maybe no elements in that magnitude
            T = np.max(t) * 2
            gamma = alpha - np.log(tol) / (2 * T)
            # NOTE: The correction alpha -> alpha-log(tol)/(2*T) is not in de Hoog's
            #   paper, but in Mathematica's Mathsource (NLapInv.m) implementation of
            #   inverse transforms
            nt = len(t)
            run = np.arange(2 * M + 1)  # so there are 2M+1 terms in Fourier series expansion

            # Find F argument, call F with it, get 'a' coefficients in power series
            s = gamma + 1j * np.pi * run / T
            a = F(s, *args)
            # a = a.flatten()
            a[0] = a[0] / 2  # zero term is halved

            # Build up e and q tables. Superscript is now row index, subscript column
            # CAREFUL: paper uses null index, so all indices are shifted by 1 here
            e = zer((2 * M + 1, M + 1)+a.shape[1:], dtype=complex)
            q = zer((2 * M, M + 1)+a.shape[1:], dtype=complex)  # Column 0 (here: 1) does not exist
            e[:, 0] = zer((2 * M + 1,)+a.shape[1:])
            q[:, 1] = a[1:2 * M + 1] / a[:2 * M]
            for r in range(2, M + 2):  # Step through columns (called r...)
                e[:2 * (M - r+1), r-1] = q[1:2 * (M - r+1) + 1, r-1] - q[:2 * (M - r+1), r-1] + e[1:2 * (M - r+1) + 1, r - 2]
                if r < M:  # One column fewer for q
                    rq = r + 1
                    q[:2 * (M - rq) + 2, rq-1] = q[1:2 * (M - rq) + 3, rq - 2] * e[1:2 * (M - rq) + 3, rq - 2] / e[:2 * (M - rq) + 2, rq - 2]

            # Build up d vector (index shift: 1)
            d = zer((2 * M + 1,)+a.shape[1:], dtype=complex)
            d[0] = a[0]
#             d[1::2] = -q[:M, 1:M + 1].diagonal()
#             d[2::2] = -e[:M, 1:M + 1].diagonal()
            d[1::2] = -q[0, 1:M + 1]#.T
            d[2::2] = -e[0, 1:M + 1]#.T

            # Build up A and B vectors (index shift: 2)co
            # - now make into matrices, one row for each time
            A = zer((2 * M + 2, nt)+a.shape[1:], dtype=complex)
            B = zer((2 * M + 2, nt)+a.shape[1:], dtype=complex)
            A[1, :] = d[0] * np.ones((nt,)+a.shape[1:])
            B[:2, :] = np.ones((2, nt)+a.shape[1:])
            target_shape = (len(t),) + a.shape[1:]
            reshape_dims = (len(t),) + (1,) * (len(a.shape[1:]))
            z = np.exp(1j * np.pi * t / T).reshape(reshape_dims) * np.ones(target_shape)  # Row vector
            
            
            
            dz = d[:,None] * z[None,:]

            for n in range(2, 2 * M + 2):
                A[n, :] = A[n - 1, :] + dz[n - 1, :] * A[n - 2, :]
                B[n, :] = B[n - 1, :] + dz[n - 1, :] * B[n - 2, :]

            # Double acceleration
            h2M = 0.5 * (1 + (d[2 * M - 1] - d[2 * M]) * z)
            R2Mz = -h2M * (1 - np.sqrt(1 + d[2 * M] * z / (h2M**2)))
            A[2 * M + 1, :] = A[2 * M, :] + R2Mz * A[2 * M - 1, :]
            B[2 * M + 1, :] = B[2 * M, :] + R2Mz * B[2 * M - 1, :]

            # Inversion, vectorized for times, make result a column vector
            fpiece = (1 / T * (np.exp(gamma * t).reshape(reshape_dims) * np.ones(target_shape)) * np.real(A[2 * M + 1, :] / B[2 * M + 1, :]))
            # print(fpiece.shape())
            f.extend(fpiece)

    return np.array(f)
```

```python

U = 1.0
x_list = (xs_con_0*1000)/(stream_width/2)#np.linspace(0,5*awl,101)[1:]
D_eps = 10**-6
n_tb = 7
zs_ext = np.linspace(-1,1,n_tb+1)
delta_z = 2/n_tb
zs = (zs_ext[1:]+zs_ext[:-1])/2

def run_river_dispersion(A_scour,n_bound,beta_transp,fac_k,D_fac):
    RMSEs = np.zeros((len(x_list),n_tb,n_tb))
    print(A_scour,n_bound,beta_transp,fac_k,D_fac)
    a2 = 2*Cf*beta*chi
    # if meandering[i_A]:
    a3 = -chi
    a4 = Cf*beta*chi**2*((1+5*np.sqrt(Cf))*(A_scour+chi**3*Fr**2)+1)
    # else:
    #     a3 = a4 = 0.0
    # plt.plot(ss,thetas)
    # plt.plot(ss,thetas_sin_cos_ser(b_and_a))
    vel_grad_curv_cos = lambda s, a2, a3, a4: np.sum((a3*(-A_i[:,None]*2*np.pi/lam_i[:,None]*np.cos(2*np.pi*s/lam_i[:,None]+phi_i[:,None]))
                    +a4*(a2*(-A_i[:,None]*2*np.pi/lam_i[:,None]*np.cos(2*np.pi*s/lam_i[:,None]+phi_i[:,None]))
                          + 2*np.pi/lam_i[:,None]*(-A_i[:,None]*2*np.pi/lam_i[:,None]*np.sin(2*np.pi*s/lam_i[:,None]+phi_i[:,None])))
                           /(a2**2+(2 * np.pi / lam_i[:,None])**2)),axis = 0)
    # plt.plot(ss,curv_cos(ss))
    ss_meas = release_s+xs_con_0*1000/(stream_width/2)
    
    epsilon = 0.01
    vel_grad = vel_grad_curv_cos(ss, a2, a3, a4)    
    vel_bank_left = np.minimum(np.maximum(1+(vel_grad[1:]+vel_grad[:-1])/2, np.ones(len(ss)-1)*epsilon),np.ones(len(ss)-1)*2.0)
    vel_bank_right = np.minimum(np.maximum(1-(vel_grad[1:]+vel_grad[:-1])/2, np.ones(len(ss)-1)*epsilon),np.ones(len(ss)-1)*2.0)
    
    
    
    xs_bank_left, ys_bank_left = offset_curve(xs,ys, 1.0)
    xs_bank_right, ys_bank_right = offset_curve(xs,ys, -1.0)
    
    # Create segments and corresponding LineCollections
    seg1 = segments_from_xy(xs_bank_left, ys_bank_left)
    seg2 = segments_from_xy(xs_bank_right, ys_bank_right)
    
    # norm = Normalize(vmin=max(1+min(vel_grad),epsilon), vmax=min(1+max(vel_grad),2))
    # lc1 = LineCollection(seg1, cmap='coolwarm', norm=norm)
    # lc1.set_array(vel_bank_left)
    # lc1.set_linewidth(2)
    
    # lc2 = LineCollection(seg2, cmap='coolwarm', norm=norm)
    # lc2.set_array(vel_bank_right)
    # lc2.set_linewidth(2)
    
    # fig, ax = plt.subplots(figsize = (20,20))
    # ax.plot(xs, ys, c='r', linestyle='dashed')
    # ax.add_collection(lc1)
    # ax.add_collection(lc2)
    
    # ax.set_aspect('equal', adjustable='box')
    # ax.set_ylim(min(ys)-2.0,max(ys)+2.0)
    # # ax.set_title('Seed='+str(seed)+r', $\beta$='+str(round(beta,1))+r', $\theta$='+str(round(theta_0_deg,1))+r'$^{\circ}$, $J_s$='+str(round(Js,3))+', $J_s$='+str(round(Jf,3))+', $Fr$='+str(round(Fr,3))+', $C_{f}$='+str(round(Cf0,3)))
    # plt.colorbar(lc1, ax=ax, label='Velocity', orientation='horizontal', fraction=0.05, pad=0.02,  aspect=120)  # Optional: show color scale
    # meas_indices = np.argmin(np.abs(ss_meas-ss[:,None]),axis = 0)
    # ax.scatter(xs[meas_indices], ys[meas_indices],marker='^',c = 'tab:orange', s = 100,zorder = 3)
    # plt.show()
    # plt.plot((ss[1:]+ss[:-1])/2*2*stream_width/1000,vel_bank_left,label='Left bank')
    # plt.plot((ss[1:]+ss[:-1])/2*2*stream_width/1000,vel_bank_right,label='Right bank')
    # plt.xlabel('x [km]')
    # plt.ylabel('u/U')
    # plt.legend()
    # plt.show()
    v_bound = lambda z: 1-np.abs(z)**(n_bound-1)

    vs = lambda x: np.minimum(np.maximum((v_bound(zs)*(zs*vel_grad_curv_cos(x+release_s,a2, a3, a4)[:,None]+1)),np.ones((len(x),n_tb))*epsilon),np.ones((len(x),n_tb))*2.0)
    vs_ext = lambda x: np.minimum(np.maximum((v_bound(zs_ext)*(zs_ext*vel_grad_curv_cos(x+release_s,a2, a3, a4)[:,None]+1)),np.ones((len(x),n_tb+1))*epsilon),np.ones((len(x),n_tb+1))*2.0)
    
    def V_f(x):
        V = zer((len(x),n_tb,n_tb))
        V[:,np.arange(n_tb),np.arange(n_tb)] = vs(x)
        return V
    def Q_f(x):
        u_stars = vs_ext(x)*np.sqrt(Cf)
        Ds = D_fac*u_stars*1/beta
        Q = zer((len(x),n_tb,n_tb))
        Q[:,np.arange(n_tb-1),np.arange(n_tb-1)+1] = 1/delta_z**2*Ds[:,1:-1]
        Q[:,np.arange(1,n_tb),np.arange(1,n_tb)-1] = 1/delta_z**2*Ds[:,1:-1]
        Q[:,np.arange(n_tb),np.arange(n_tb)] = -1/delta_z**2*(Ds[:,1:]+Ds[:,:-1])
        return Q
    def D_f_inv(x):
        u_stars = vs(x)*np.sqrt(Cf)
        Ds = D_fac*u_stars*1/beta
        D_inv = zer((len(x),n_tb,n_tb))
        D_inv[:,np.arange(n_tb),np.arange(n_tb)] = 1/Ds
        return D_inv
    def D_eps_f_inv(x):
        Ds = D_eps
        D_inv = zer((len(x),n_tb,n_tb))
        D_inv[:,np.arange(n_tb),np.arange(n_tb)] = 1/Ds
        return D_inv
    def K_f(x):
        u_stars = vs(x)*np.sqrt(Cf)
        Ds = 0.15*u_stars*1/beta
        K = zer((len(x),n_tb,n_tb))
        K[:,np.arange(n_tb),np.arange(n_tb)] = Ds*beta**2*fac_k #D/h^2
        return K
    A_f = lambda x: conc((conc((zer((len(x),n_tb,n_tb)), np.ones(len(x))[:,None,None]*np.eye(n_tb)[None,:,:],zer((len(x),n_tb,n_tb)),zer((len(x),n_tb,n_tb))),axis = 1),
                                       conc((D_f_inv(x)@(-Q_f(x)+beta_transp*K_f(x)), D_f_inv(x)@V_f(x), -beta_transp*D_f_inv(x)@K_f(x), zer((len(x),n_tb,n_tb))),axis = 1), 
                                       conc((zer((len(x),n_tb,n_tb)), zer((len(x),n_tb,n_tb)), zer((len(x),n_tb,n_tb)), np.ones(len(x))[:,None,None]*np.eye(n_tb)[None,:,:]),axis = 1),
                                       conc((-D_eps_f_inv(x)@K_f(x), zer((len(x),n_tb,n_tb)),D_eps_f_inv(x)@K_f(x), zer((len(x),n_tb,n_tb))),axis = 1)),axis = 2)
    B_f = lambda x: conc((conc((zer((len(x),n_tb,n_tb)),zer((len(x),n_tb,n_tb)), zer((len(x),n_tb,n_tb)),zer((len(x),n_tb,n_tb))),axis = 1),
                                   conc((D_f_inv(x), zer((len(x),n_tb,n_tb)), zer((len(x),n_tb,n_tb)), zer((len(x),n_tb,n_tb))),axis = 1), 
                                   conc((zer((len(x),n_tb,n_tb)), zer((len(x),n_tb,n_tb)), zer((len(x),n_tb,n_tb)), zer((len(x),n_tb,n_tb))),axis = 1),
                                   conc((zer((len(x),n_tb,n_tb)), zer((len(x),n_tb,n_tb)), D_eps_f_inv(x), zer((len(x),n_tb,n_tb))),axis = 1)),axis = 2)
                
    m1s = zer((len(x_list),n_tb,n_tb))
    m2s = zer((len(x_list),n_tb,n_tb))
    m3s = zer((len(x_list),n_tb,n_tb))
    
    btcs = []
    for x_i,x in enumerate(x_list):
        ts = np.linspace(x/U*0.5,x/U*7.0,976)[1:]
        x_s = np.linspace(0,x,500)
        A = A_f(x_s)
        B = B_f(x_s)
        A1 = np.trapz(A,x_s,axis = 0)
        B1 = np.trapz(B,x_s,axis = 0)
        # A2 = np.trapz(np.trapz(np.matmul(A[None,:,:,:],A[:,None,:,:])-np.matmul(A[:,None,:,:],A[None,:,:,:]),x_s,axis = 0),x_s,axis = 0)
        # AB = np.trapz(np.trapz(np.matmul(A[None,:,:,:],B[:,None,:,:])+np.matmul(B[None,:,:,:],A[:,None,:,:])-
        #                        np.matmul(A[:,None,:,:],B[None,:,:,:])-np.matmul(B[:,None,:,:],A[None,:,:,:]),x_s,axis = 0),x_s,axis = 0)
        # B2 = np.trapz(np.trapz(np.matmul(B[None,:,:,:],B[:,None,:,:])-np.matmul(B[:,None,:,:],B[None,:,:,:]),x_s,axis = 0),x_s,axis = 0)    
        def c_lap_space(s):
            result = zer((len(s),n_tb,n_tb))
            # result = zer((len(s)))
            for i in range(len(s)):
                M = (A1+s[i]*B1)#+(A2+s[i]*AB+B2*s[i]**2)
                lambdas, S = np.linalg.eig(M)
                exp_lambdas = np.exp(lambdas)
                exp_lambdas[np.isnan(exp_lambdas)|np.isinf(exp_lambdas)] = 0.0
                exp_M = S@np.diag(exp_lambdas)@np.linalg.inv(S)#sp.linalg.expm(M)#
                result[i] = exp_M[:n_tb,:n_tb]
                # result[i] = np.sum(exp_M@(np.eye(n_tb)[int(pos*n_tb)]))
            return result
        # c_lap_space_vec = np.vectorize(c_lap_space)
        btc = invlap(c_lap_space,ts,[])
        btc[np.isnan(btc)] = 0.0
        btcs.append(btc)
        for i in range(n_tb):
            for j in range(n_tb):
                m1s[x_i,i,j] = np.trapz(btc[:,i,j]*ts,ts)/np.trapz(btc[:,i,j],ts)
                m2s[x_i,i,j] = np.trapz(btc[:,i,j]*(ts-m1s[x_i,i,j])**2,ts)/np.trapz(btc[:,i,j],ts)
                m3s[x_i,i,j] = np.trapz(btc[:,i,j]*(ts-m1s[x_i,i,j])**3,ts)/np.trapz(btc[:,i,j],ts)

                t_fac = ((np.trapz(btcs_con_0[x_i]*ts_con_0[x_i],ts_con_0[x_i])/np.trapz(btcs_con_0[x_i],ts_con_0[x_i]))
                         /(np.trapz(btc[:,i,j]*ts,ts)/np.trapz(btc[:,i,j],ts)))
                btc_interp = np.interp(ts_con_0[x_i]/t_fac,ts,btc[:,i,j]/np.trapz(btc[:,i,j],ts))
                RMSEs[x_i,i,j] = RMSE_func(ts_con_0[x_i]/t_fac,btcs_con_0[x_i]/np.trapz(btcs_con_0[x_i],ts_con_0[x_i]/t_fac),btc_interp)
    print(RMSEs[:,n_tb//2,n_tb//2])
    COV = m2s**(1/2)/m1s
    CSK = m3s/m2s**(3/2)
    print(COV[:,n_tb//2,n_tb//2])
    print(CSK[:,n_tb//2,n_tb//2])
    print()
    return RMSEs,COV,CSK,btcs
```

## 3D model 2

```python
U = 1.0
x_list = (xs_con_0*1000)/(stream_width/2)#np.linspace(0,5*awl,101)[1:]
D_eps = 10**-6
n_tb = 7
zs_ext = np.linspace(-1,1,n_tb+1)
delta_z = 2/n_tb
zs = (zs_ext[1:]+zs_ext[:-1])/2

def run_river_dispersion(A_scour,n_bound,beta_transp,fac_k,D_fac):
    RMSEs = np.zeros((len(x_list),n_tb,n_tb))
    print(A_scour,n_bound,beta_transp,fac_k,D_fac)
    a2 = 2*Cf*beta*chi
    # if meandering[i_A]:
    a3 = -chi
    a4 = Cf*beta*chi**2*((1+5*np.sqrt(Cf))*(A_scour+chi**3*Fr**2)+1)
    # else:
    #     a3 = a4 = 0.0
    # plt.plot(ss,thetas)
    # plt.plot(ss,thetas_sin_cos_ser(b_and_a))
    vel_grad_curv_cos = lambda s, a2, a3, a4: np.sum((a3*(-A_i[:,None]*2*np.pi/lam_i[:,None]*np.cos(2*np.pi*s/lam_i[:,None]+phi_i[:,None]))
                    +a4*(a2*(-A_i[:,None]*2*np.pi/lam_i[:,None]*np.cos(2*np.pi*s/lam_i[:,None]+phi_i[:,None]))
                          + 2*np.pi/lam_i[:,None]*(-A_i[:,None]*2*np.pi/lam_i[:,None]*np.sin(2*np.pi*s/lam_i[:,None]+phi_i[:,None])))
                           /(a2**2+(2 * np.pi / lam_i[:,None])**2)),axis = 0)
    # plt.plot(ss,curv_cos(ss))
    ss_meas = release_s+xs_con_0*1000/(stream_width/2)
    
    epsilon = 0.01
    vel_grad = vel_grad_curv_cos(ss, a2, a3, a4)    
    vel_bank_left = np.minimum(np.maximum(1+(vel_grad[1:]+vel_grad[:-1])/2, np.ones(len(ss)-1)*epsilon),np.ones(len(ss)-1)*2.0)
    vel_bank_right = np.minimum(np.maximum(1-(vel_grad[1:]+vel_grad[:-1])/2, np.ones(len(ss)-1)*epsilon),np.ones(len(ss)-1)*2.0)
    
    
    
    xs_bank_left, ys_bank_left = offset_curve(xs,ys, 1.0)
    xs_bank_right, ys_bank_right = offset_curve(xs,ys, -1.0)
    
    # Create segments and corresponding LineCollections
    seg1 = segments_from_xy(xs_bank_left, ys_bank_left)
    seg2 = segments_from_xy(xs_bank_right, ys_bank_right)
    
    v_bound = lambda z: 1-np.abs(z)**(n_bound-1)

    vs = lambda x: np.minimum(np.maximum((v_bound(zs)*(zs*vel_grad_curv_cos(x+release_s,a2, a3, a4)[:,None]+1)),np.ones((len(x),n_tb))*epsilon),np.ones((len(x),n_tb))*2.0)
    vs_ext = lambda x: np.minimum(np.maximum((v_bound(zs_ext)*(zs_ext*vel_grad_curv_cos(x+release_s,a2, a3, a4)[:,None]+1)),np.ones((len(x),n_tb+1))*epsilon),np.ones((len(x),n_tb+1))*2.0)
    
    def V_f(x):
        V = zer((len(x),n_tb,n_tb))
        V[:,np.arange(n_tb),np.arange(n_tb)] = vs(x)
        return V
    def Q_f(x):
        u_stars = vs_ext(x)*np.sqrt(Cf)
        Ds = D_fac*u_stars*1/beta
        Q = zer((len(x),n_tb,n_tb))
        Q[:,np.arange(n_tb-1),np.arange(n_tb-1)+1] = 1/delta_z**2*Ds[:,1:-1]
        Q[:,np.arange(1,n_tb),np.arange(1,n_tb)-1] = 1/delta_z**2*Ds[:,1:-1]
        Q[:,np.arange(n_tb),np.arange(n_tb)] = -1/delta_z**2*(Ds[:,1:]+Ds[:,:-1])
        return Q
    def D_f_inv(x):
        u_stars = vs(x)*np.sqrt(Cf)
        Ds = D_fac*u_stars*1/beta
        D_inv = zer((len(x),n_tb,n_tb))
        D_inv[:,np.arange(n_tb),np.arange(n_tb)] = 1/Ds
        return D_inv
    def s_betag_f(x,s):
        u_stars = vs(x)*np.sqrt(Cf)
        Ds = 0.15*u_stars*1/beta
        ks = Ds*beta**2*fac_k
        
        s_betag = zer((len(s),len(x),n_tb,n_tb))
        s_betag = (s[:,None,None,None]+
                   beta_transp*ks[None,:,:,None]*s[:,None,None,None]/(ks[None,:,:,None]+s[:,None,None,None]))*np.eye(n_tb)[None,None,:,:]
        return s_betag
    A_f = lambda x, s: conc((conc((zer((len(s),len(x),n_tb,n_tb)), np.ones((len(s),len(x)))[:,:,None,None]*np.eye(n_tb)[None,None,:,:]),axis = 2),
                            conc((D_f_inv(x)[None,:,:,:]@(-Q_f(x)[None,:,:,:]+s_betag_f(x,s)), np.ones(len(s))[:,None,None,None]*(D_f_inv(x)@V_f(x))),axis = 2)),axis = 3)
    # A_f = lambda x, s: conc((conc((zer((len(s),len(x),n_tb,n_tb)), np.ones((len(s),len(x)))[:,:,None,None]*np.eye(n_tb)[None,None,:,:]),axis = 2),
    #                         conc((D_f_inv(x)[None,:,:,:]@(-np.ones(len(s))[:,None,None,None]*Q_f(x)[None,:,:,:]), np.ones(len(s))[:,None,None,None]*(D_f_inv(x)@V_f(x))),axis = 2)),axis = 3)

    m1s = zer((len(x_list),n_tb,n_tb))
    m2s = zer((len(x_list),n_tb,n_tb))
    m3s = zer((len(x_list),n_tb,n_tb))
    
    btcs = []
    for x_i,x in enumerate(x_list):
        ts = np.linspace(x/U*0.5,x/U*3.0,376)[1:]
        x_s = np.linspace(0,x,250)
        def c_lap_space(s):
            result = zer((len(s),n_tb,n_tb))
            A = A_f(x_s,s)
            M = np.trapz(A,x_s,axis = 1) 
            lambdas, S = np.linalg.eig(M)
            exp_lambdas = np.exp(lambdas)
            exp_lambdas[np.isnan(exp_lambdas)|np.isinf(exp_lambdas)] = 0.0
            result = S@(np.eye(2*n_tb)*exp_lambdas[:,:,None])@np.linalg.inv(S)
            
            # for i in range(len(s)):
            #     exp_M = S[i]@np.diag(exp_lambdas[i])@np.linalg.inv(S[i])
            #     result[i] = exp_M[:n_tb,:n_tb]
            return result
        btc = invlap(c_lap_space,ts,[])
        btc[np.isnan(btc)] = 0.0
        btcs.append(btc)
        for i in range(n_tb):
            for j in range(n_tb):
                m1s[x_i,i,j] = np.trapz(btc[:,i,j]*ts,ts)/np.trapz(btc[:,i,j],ts)
                m2s[x_i,i,j] = np.trapz(btc[:,i,j]*(ts-m1s[x_i,i,j])**2,ts)/np.trapz(btc[:,i,j],ts)
                m3s[x_i,i,j] = np.trapz(btc[:,i,j]*(ts-m1s[x_i,i,j])**3,ts)/np.trapz(btc[:,i,j],ts)

                t_fac = ((np.trapz(btcs_con_0[x_i]*ts_con_0[x_i],ts_con_0[x_i])/np.trapz(btcs_con_0[x_i],ts_con_0[x_i]))
                         /(np.trapz(btc[:,i,j]*ts,ts)/np.trapz(btc[:,i,j],ts)))
                btc_interp = np.interp(ts_con_0[x_i]/t_fac,ts,btc[:,i,j]/np.trapz(btc[:,i,j],ts))
                RMSEs[x_i,i,j] = RMSE_func(ts_con_0[x_i]/t_fac,btcs_con_0[x_i]/np.trapz(btcs_con_0[x_i],ts_con_0[x_i]/t_fac),btc_interp)
    print(RMSEs[:,n_tb//2,n_tb//2])
    
    COV = m2s**(1/2)/m1s
    CSK = m3s/m2s**(3/2)
    print(COV[:,n_tb//2,n_tb//2])
    print(CSK[:,n_tb//2,n_tb//2])
    print()
    return RMSEs,COV,CSK,btcs
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import roots_laguerre



# Define the function to approximate
def exact_ftion(K, S):
    return K * S / (K + S)

# Separable approximation using outer products and sum
def approx_ftion(k_vals, s_vals, N):
    # Get Gauss-Laguerre nodes (t_i) and weights (w_i)
    t_i, w_i = roots_laguerre(N)  # both are shape (N,)
    k_vals = np.atleast_1d(k_vals)
    s_vals = np.atleast_1d(s_vals)

    K = k_vals[:, np.newaxis]  # Shape (len(k), 1)
    S = s_vals[np.newaxis, :]  # Shape (1, len(s))

    result = zer((len(k_vals), len(s_vals)))+K

    # Loop over quadrature points
    for i in range(N):
        f_i = w_i[i]*10 * K**2 * np.exp(-K * t_i[i]*10)  # Shape (len(k), 1)
        g_i = np.exp(-S * t_i[i]*10)           # Shape (1, len(s))
        result -= f_i * g_i                     # Outer product shape (len(k), len(s))

    return result

# Evaluation grid
k_vals = np.linspace(0.1, 10, 100)
s_vals = np.linspace(0.1, 10, 100)
K, S = np.meshgrid(k_vals, s_vals)  # For plotting

# Compute exact and approximate values
exact = exact_ftion(K, S)
approx = approx_ftion(k_vals, s_vals,100)

# Plot relative error
rel_error = np.abs(approx - exact) / (np.abs(exact) + 1e-12)

plt.figure(figsize=(8, 6))
plt.contourf(K, S, rel_error, levels=100, cmap='viridis')
plt.colorbar(label='Relative Error')
plt.xlabel('k')
plt.ylabel('s')
plt.title('Relative Error of Separable Approximation')
plt.tight_layout()
plt.show()
```
